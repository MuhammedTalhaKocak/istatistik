### T Dağılım Ortalama Güven Aralığı
T Dağılımı, örneklem sayısı az olduğunda (n<30) veya ana kütle standart sapması bilinmediğinde popülasyon ortalamasını tahmin etmek için kullanılır.
Formül: $$\boxed{\bar{x} - t_{\alpha/2, \, n-1} \cdot \frac{s}{\sqrt{n}} \le \mu \le \bar{x} + t_{\alpha/2, \, n-1} \cdot \frac{s}{\sqrt{n}}}$$

- t değeri **serbestlik derecesine** göre değişir.
- Python'da stats.t.ppf() veya stats.t.interval() kullanılır (norm yerine t modülü).

```python
import numpy as np
from scipy import stats
veri = np.array([90,85,76,83,89,92,79,81,89,84,97,72,77,90,100,69,80,94,87,86])
ortalama = np.mean(veri)
standart_sapma = np.std(veri, ddof=1) # elimizde bir örneklem var o yüzden bu örnekleme ait sapmanın serbestlik derecesi 1
standart_hata = (standart_sapma) / (len(veri) ** 0.5)
t_alt_sinir = stats.t.ppf(q = 0.025, df = len(veri) - 1)
t_ust_sinir = stats.t.ppf(q = 0.975, df = len(veri) - 1)
t_alt_sinir_99 = stats.t.ppf(q = 0.005, df = len(veri) - 1)
t_ust_sinir_99 = stats.t.ppf(q = 0.995, df = len(veri) - 1)
print(f"%95 güven düzeyiyle aralığımız {ortalama + (t_alt_sinir * standart_hata)} ile {ortalama + (t_ust_sinir * standart_hata)} arasındadır.")
print(f"SciPy'ın hesabı ile %95 güven aralığı: {stats.t.interval(confidence=0.95, df = len(veri) - 1, loc = ortalama, scale = standart_hata)}")
print(f"%99 güven düzeyiyle aralığımız {ortalama + (t_alt_sinir_99 * standart_hata)} ile {ortalama + (t_ust_sinir_99 * standart_hata)} arasındadır.")
print(f"SciPy'ın hesabı ile %99 güven aralığı: {stats.t.interval(confidence=0.99, df = len(veri) - 1, loc = ortalama, scale = standart_hata)}")
```
