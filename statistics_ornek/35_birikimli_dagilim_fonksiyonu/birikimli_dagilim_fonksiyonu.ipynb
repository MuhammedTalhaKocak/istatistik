# 35) Birikimli Dağılım Fonksiyonu
Birikimli dağılım fonksiyonu (kümülatif dağılım fonksiyonu veya CDF), olasılık ve istatistikte bir rassal değişkenin belirli bir değerden küçük veya o değere eşit olma olasılığını veren temel bir fonksiyondur. 

CDF, kesikli değişkenlerde PMF'in birikmiş toplamıdır:
$$F(x) = P(X \le x) = \sum_{x_i \le x} P(x_i)$$

Sürekli değişkenlerde ise PDF'in birikmiş integralidir:
$$F(x) = P(X \le x) = \int_{-\infty}^{x} f(t)\, dt$$

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# --- 1. Kesikli CDF - Binom Dağılımı ---
n, p = 10, 0.5
x_cdf_binom = np.arange(0, n + 1)
y_cdf_binom = stats.binom.cdf(x_cdf_binom, n, p)

# CDF kesiklide "basamak" (step) şeklinde çizilir çünkü olasılık sadece 
# tam sayı noktalarında birikir, aralarda sabit kalır
ax1.step(x_cdf_binom, y_cdf_binom, where='post', color='C0', linewidth=2)
ax1.scatter(x_cdf_binom, y_cdf_binom, color='C0', zorder=5)
ax1.set_title("Birikimli Dağılım Fonksiyonu (CDF)\nBinom Dağılımı (10 Para Atışı)", fontsize=11)
ax1.set_xlabel("Gelen Tura Sayısı (x)")
ax1.set_ylabel("P(X ≤ x)")
ax1.set_xticks(x_cdf_binom)
ax1.grid(True, linestyle="--", alpha=0.5)

# --- 2. Sürekli CDF - Normal Dağılım ---
mu, sigma = 0, 1
x_cdf_norm = np.linspace(-4, 4, 500)
y_cdf_norm = stats.norm.cdf(x_cdf_norm, mu, sigma)

# CDF süreklide düz, kesintisiz bir S-eğrisi (sigmoid benzeri) şeklindedir
ax2.plot(x_cdf_norm, y_cdf_norm, 'C1-', linewidth=2)
ax2.axhline(y=0.5, color='gray', linestyle=':', alpha=0.7)
ax2.axvline(x=0, color='gray', linestyle=':', alpha=0.7)
ax2.set_title("Birikimli Dağılım Fonksiyonu (CDF)\nStandart Normal Dağılım", fontsize=11)
ax2.set_xlabel("Değer (x)")
ax2.set_ylabel("P(X ≤ x)")
ax2.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

### Sonuç
Soldaki grafik 0'dan başlayarak her bir değerin kendisine eşit veya kendisinden küçük olma olasılığını kümülatif olarak verirken (örneğin 10 madeni para atışında 10 tura veya daha az tura gelme olasılığı = 1 çünkü 10 madeni para atışında en fazla 10 kere tura gelebilir, bu yüzden 10 veya daha az tura gelme olasılığı kesinlikle gerçekleşir.) sağdaki grafik ise standart bir normal dağılımda x değerinin veya x değerinden daha az bir değerin gelme olasılığını kümülatif olarak verir. Örneğin X=0 veya daha küçük bir değer gelme olasılığı %50'dir çünkü standart normal dağılım, kendi ortalaması olan 0 etrafında simetriktir. Bu simetri sayesinde dağılımın tam yarısı 0'ın solunda, yarısı sağında kalır.
