# Kategorik Değişken Kodlama
Kategorik Değişken Kodlama, metin/etiket formatındaki kategorik değişkenleri (örn: "Kırmızı", "Mavi", "Yeşil"), çoğu istatistiksel/ML 
algoritmasının anlayabileceği **sayısal forma** çevirme işlemidir. Çoğu algoritma, metin verisiyle doğrudan çalışamaz.

## İki Ana Yaklaşım (Ölçek Türüne Göre — Konu 4'ü Hatırla)

### 1. Nominal Değişkenler İçin (Sıralama Yok)
Kategoriler arasında **hiçbir doğal sıralama** yoksa (örn: Renk, Şehir, Kategori), **One-Hot Encoding** kullanılır. Her kategori için ayrı bir 0/1 sütunu oluşturulur.

### 2. Ordinal Değişkenler İçin (Sıralama Var)
Kategoriler arasında **doğal bir sıralama** varsa (örn: "Düşük/Orta/Yüksek", "İlkokul/Lise/Üniversite"), sıralamayı koruyan bir **sayısal etiketleme** (Label Encoding) yapılır (Ordinal Değişken Dönüşümü).

## Basit Label Encoding (Genel Amaçlı)
```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Kategori_Kodlanmis'] = le.fit_transform(df['Kategori'])
```

## ÖNEMLİ UYARI — Nominal Değişkende Label Encoding Kullanmanın Tehlikesi
Eğer **nominal** bir değişkene (örn: Renk: Kırmızı, Mavi, Yeşil) Label Encoding uygularsak, bu kategoriler **rastgele sayılara** (0, 1, 2 gibi) dönüşür — ama algoritma bu sayıları **"Yeşil (2), Kırmızı'dan (0) büyük"** gibi **yanlış bir sıralama anlamı** ile yorumlayabilir! Bu, "nominal ölçekte sıralama yoktur" kuralının ihlalidir. Bu yüzden nominal değişkenlerde **One-Hot Encoding tercih 
edilmelidir**, Label Encoding değil.

## Python'da Kullanımı
```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({'Renk': ['Kırmızı', 'Mavi', 'Yeşil', 'Kırmızı', 'Mavi']})

le = LabelEncoder()
df['Renk_Kodlanmis'] = le.fit_transform(df['Renk'])
print(df)
```

````python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({'Renk': ['Kırmızı', 'Mavi', 'Yeşil', 'Kırmızı', 'Mavi', 'Yeşil']})

le = LabelEncoder()
df['Renk_Kodlanmis'] = le.fit_transform(df['Renk'])

print(df)
print("\nHangi sayı hangi renge karşılık geliyor:")
for i, sinif in enumerate(le.classes_):
    print(f"{sinif} -> {i}")
````

Yukarıdaki sonuçta görüldüğü üzere renkler arasında bir sıralama farkı olmamasına rağmen Label Encoding yüzünden Makine Öğrenmesi Modelleri Yeşil > Mavi > Kırmızı sanabilir. Bu da yanlış bir durumdur, bu sorunu bir sonraki dosyada One-Hot Encoding ile çözeceğiz. Yeniden hatırlatmak gerekirse, Label Encoding değişkenler arasında bir sıralama, hiyerarşi varsa kullanılır.
