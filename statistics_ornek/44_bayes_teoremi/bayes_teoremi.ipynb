# 44) Bayes Teoremi

Bayes Teoremi, Koşullu Olasılığın (Konu 43) **yönünü tersine çevirmemizi** sağlayan bir formüldür. Elimizde $P(B|A)$ varsa (ya da bunu hesaplamak $P(A|B)$'den daha kolaysa), Bayes Teoremi ile $P(A|B)$'yi bulabiliriz.

## Formül
$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$

### Formülün Parçaları (İsimlendirme)
- **$P(A)$ Prior (Önsel Olasılık):** Herhangi bir yeni bilgi (B) öğrenmeden ÖNCE, A hakkında bildiğimiz olasılık.
- **$P(B|A)$ Likelihood (Olabilirlik):** A doğruyken, B'nin gözlenme olasılığı.
- **$P(A|B)$ Posterior (Sonsal Olasılık):** B'yi gözlemledikten SONRA, A hakkındaki güncellenmiş olasılığımız.
- **$P(B)$ Evidence (Kanıt):** B'nin genel/toplam gerçekleşme olasılığı.

**Bayes Teoremi'nin özü:** "Yeni bir kanıt (B) gördüğümüzde, önceki inancımızı (Prior) bu kanıtla güncelleyip, yeni bir inanca (Posterior) ulaşırız."

## Klasik Örnek — Hastalık Testi (Yanlış Pozitif Tuzağı)
Nadir bir hastalık toplumda %1 görülüyor ($P(\text{Hasta})=0.01$). Bir test, hastaları %95 doğrulukla tespit ediyor ($P(\text{Test+}|\text{Hasta})=0.95$), ama sağlıklı birine de %10 ihtimalle yanlış pozitif verebiliyor.
($P(\text{Test+}|\text{Sağlıklı})=0.10$).

**Soru:** Testi pozitif çıkan birinin, gerçekten hasta olma olasılığı nedir?

**Yaygın (yanlış) sezgi:** "%95 doğru test pozitif çıktı, demek ki %95 hastayım." **Bu yanlıştır** çünkü hastalık zaten çok nadir olduğu için, pozitif çıkanların büyük bir kısmı aslında **yanlış pozitiftir** (sağlıklı ama test hata yapmış). Doğru sonuç, Bayes formülüyle hesaplandığında sezgisel beklentiden çok daha düşük çıkar.

## Neden Önemli?
- **Tıp/test yorumlama:** Nadir hastalıklarda pozitif test sonuçlarını doğru yorumlamak için şart.
- **Spam filtreleme:** "Bu email'de şu kelimeler var, spam olma olasılığı nedir?" Naive Bayes sınıflandırıcısının temelidir.
- **Pazarlama/Churn tahmini:** "Bu müşteri şu davranışları gösterdi, churn etme olasılığı nedir?" gibi güncellenen tahminlerde kullanılır.
- **A/B testte Bayesian yaklaşım:** "Bayesian A/B Testi" konusunun temelini oluşturur. Sabit p-değeri yerine, sürekli güncellenen olasılıklarla karar vermeyi sağlar.

```python
# Kalite Kontrol Örneği - Bayes Teoremi Python Uygulaması

# --- Verilen Bilgiler (Prior ve Likelihood) ---
p_makine1 = 0.60      # P(Makine 1) - Prior
p_makine2 = 0.40      # P(Makine 2) - Prior
p_hatali_makine1 = 0.02   # P(Hatalı | Makine 1) - Likelihood
p_hatali_makine2 = 0.05   # P(Hatalı | Makine 2) - Likelihood

# --- Adım 1: Kesişimleri hesapla ---
p_hatali_ve_makine1 = p_hatali_makine1 * p_makine1
p_hatali_ve_makine2 = p_hatali_makine2 * p_makine2

print(f"P(Hatalı ve Makine 1) = {p_hatali_ve_makine1}")
print(f"P(Hatalı ve Makine 2) = {p_hatali_ve_makine2}")

# --- Adım 2: Toplam hatalı olasılığı (payda) ---
p_hatali = p_hatali_ve_makine1 + p_hatali_ve_makine2
print(f"\nP(Hatalı) = {p_hatali}")

# --- Adım 3: Bayes Teoremi - P(Makine 2 | Hatalı) ---
p_makine2_verildi_hatali = p_hatali_ve_makine2 / p_hatali
print(f"\nP(Makine 2 | Hatalı) = {p_makine2_verildi_hatali}")

# --- Doğrulama: Elle bulduğumuz sonuçla (20/32 = 0.625) karşılaştır ---
print(f"\nElle bulduğumuz sonuç: {20/32}")
print(f"Python'un bulduğu sonuç: {p_makine2_verildi_hatali}")
print(f"İkisi eşit mi? {abs((20/32) - p_makine2_verildi_hatali) < 0.0001}")
```
