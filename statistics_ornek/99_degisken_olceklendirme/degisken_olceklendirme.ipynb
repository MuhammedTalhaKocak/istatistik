# Değişken Ölçeklendirme (Feature Scaling)
Değişken Ölçeklendirme, farklı **birimlerde/büyüklüklerde** olan sayısal 
değişkenleri, ortak bir ölçeğe getirme işlemidir. Bu, Veri Önişleme 
bölümünün ilk ve en temel adımlarından biri.

## Neden Gerekli?
Diyelim veri setinde `Yas` (18-80 arası) ve `Gelir` (2000-50000 arası) 
var. Bu iki değişken **çok farklı büyüklüklerde** — bu, birçok algoritma 
için sorun yaratır:
- **Mesafe tabanlı yöntemler** (KNN, Mahalanobis Uzaklığı, Kümeleme algoritmaları): Gelir'in büyük sayıları, mesafe hesabına Yaş'tan çok daha fazla "ağırlık" katar, oysa ikisi de eşit önemde olabilir.
- **Regresyon/Gradient Descent tabanlı modeller**: Farklı ölçekler, öğrenme sürecini yavaşlatabilir veya bozabilir.
- **PCA gibi varyansa dayalı yöntemler:** Büyük ölçekli değişken, varyansı yapay olarak domine eder.

## İki Ana Yöntem

### 1. Min-Max Normalizasyon (Normalization)
$$x_{yeni} = \frac{x - x_{min}}{x_{max} - x_{min}}$$
Değeri **0 ile 1 arasına** sıkıştırır. Aykırı değerlere **duyarlıdır** 
(tek bir uç değer, tüm ölçeği bozabilir — Konu 85'deki Pearson'ın aykırı 
değer hassasiyetine benzer bir mantık).

### 2. Z-Score Standardizasyonu (Standardization)
$$x_{yeni} = \frac{x - \bar{x}}{s}$$
Bu, Konu 20'de öğrendiğimiz **Z-Score formülünün ta kendisi!** Sonuç, 
ortalaması 0, standart sapması 1 olan bir dağılım olur. Aykırı 
değerlere Min-Max'tan **daha az duyarlıdır** (ama tamamen bağışık 
değildir).

## Hangisi Ne Zaman?
- **Min-Max:** Değişkenin belirli bir aralıkta (0-1 gibi) olması 
  gerektiğinde (örn: görüntü işleme, bazı sinir ağı girdileri)
- **Z-Score:** Genel amaçlı, çoğu istatistiksel/ML uygulamasında 
  varsayılan tercih; veri normal dağılıma yakınsa özellikle uygun

## Python'da Kullanımı
```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Min-Max
minmax = MinMaxScaler()
df['Gelir_MinMax'] = minmax.fit_transform(df[['Gelir']])

# Z-Score (Standardization)
standart = StandardScaler()
df['Gelir_Standart'] = standart.fit_transform(df[['Gelir']])
```

````python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

np.random.seed(42)

df = pd.DataFrame({
    'Yas': np.random.randint(18, 80, 10),
    'Gelir': np.random.randint(2000, 50000, 10)
})

print("Orijinal veri:")
print(df)

# Min-Max Normalizasyon
minmax = MinMaxScaler()
df[['Yas_MinMax', 'Gelir_MinMax']] = minmax.fit_transform(df[['Yas', 'Gelir']])

# Z-Score Standardizasyon
standart = StandardScaler()
df[['Yas_Standart', 'Gelir_Standart']] = standart.fit_transform(df[['Yas', 'Gelir']])

print("\nÖlçeklendirme sonrası:")
print(df[['Yas', 'Yas_MinMax', 'Yas_Standart', 'Gelir', 'Gelir_MinMax', 'Gelir_Standart']])
````

### Sonuç
Min-Max Scaler, formülü doğrudan minimum ve maksimum değerlere dayandığı 
için aykırı değerlere karşı oldukça duyarlıdır — tek bir uç değer, tüm 
ölçeklendirmeyi bozabilir. Standard Scaler (Z-Score) ise ortalama ve 
standart sapma kullandığı için aykırı değerlere görece daha az duyarlıdır, 
bu nedenle veride aykırı değer varsa Standard Scaler tercih edilmesi 
daha güvenlidir. Bizim örneğimizde Min-Max Scaler, verinin dağılım 
şeklini bozmadan değerleri 0-1 arasına sıkıştırırken, Standard Scaler 
ortalaması 0, standart sapması 1 olan bir dağılım üretmiştir. Standard 
Scaler'ın sonuçları normal dağılımda daha anlamlı yorumlanabilse de, 
kullanımı normal dağılım şartına bağlı değildir.
