# Aykırı Değer Silme Yöntemi
En basit yaklaşım — tespit edilen aykırı değerleri **veri setinden tamamen çıkarmak.**

```python
df_temiz = df[~df['Aykiri_IQR']]
print(f"Önce: {len(df)} satır, Sonra: {len(df_temiz)} satır")
```

**Ne zaman kullanılır:** Aykırı değerin **veri hatası** olduğuna eminsek ve veri kaybı azsa.

**Dezavantajı:** Eğer aykırı değer **gerçek/önemli bir bilgi** taşıyorsa 
(örn: gerçekten çok yüksek harcama yapan VIP müşteri), onu silmek 
**değerli bilgiyi kaybetmek** demektir — özellikle churn/fraud gibi 
"nadir ama kritik" olayları incelerken bu çok tehlikelidir.

````python
import numpy as np
import pandas as pd
notlar = np.random.randint(0,100,95)
aykirilar = np.array([110,130,150,170,190])
df_puan = pd.DataFrame({'Not':notlar})
df_aykiri = pd.DataFrame({'Not':aykirilar})
df = pd.concat([df_puan, df_aykiri], ignore_index=True)
````

````python
print(df.shape)
df.head()
````

````python
df['Not'].describe()
````

````python
Q1 = df['Not'].quantile(0.25)
Q3 = df['Not'].quantile(0.75)
IQR = Q3 - Q1
alt_sinir = Q1 - (1.5*IQR) 
ust_sinir = Q3 + (1.5*IQR)
df['Aykiri_Deger_Mi'] = (df['Not'] < alt_sinir) | (df['Not'] > ust_sinir)
````

````python
df[df['Aykiri_Deger_Mi']==True]
````

IQR yöntemiyle bulduğumuz aykiri değerlerin indeksini alıp dataframe'den sileceğiz.

````python
print(df.shape)
aykiri_indeksler = df[df['Aykiri_Deger_Mi']==True].index
df = df.drop(index=aykiri_indeksler).copy()
df[df['Aykiri_Deger_Mi']==True]
print(df.shape)
````

````python
df[df['Aykiri_Deger_Mi']==True]
````

### Sonuç 
IQR yöntemiyle 150,170 ve 190 değerlerini Aykırı olarak tespit edip bu değerlerin indekslerini bir değişkene atayarak bu değişken üzerinden `df.drop(index=aykiri_indeksler)` komutunu çalıştırarak aykırı değerleri veri setinden çıkardık.
