# 1) Veri Analizi Nedir?
Veri analizi, faydalı bilgiler üretme, sonuçları anlatma, aktarma ve karar vermeyi destekleme ile verileri inceleme, temizleme, düzenleme ve dönüştürme işlemidir. Veri analizi çeşitli teknikleri içeren, işletme, pazarlama, finans, sosyal bilimler gibi farklı alanlarda kullanılan çok çeşitli yaklaşımlara sahiptir. 

## Veri Analizi Sürecinin 4 Aşaması
1. **Toplama** — Analiz edilecek verinin elde edilmesi
2. **Temizleme** — Eksik/hatalı/aykırı değerlerin düzeltilmesi
3. **Analiz** — İstatistiksel yöntemlerle veriden anlam çıkarılması
4. **Yorumlama / Sunum** — Bulunan sonuçların bir karara/aksiyona dönüştürülmesi

Sonuç olarak elimizdeki ham veriler işlenmeden bir şey ifade etmez, ham veriler çeşitli yöntem ve tekniklerle insanların anlayabileceği formata dönüştürülmelidir.
