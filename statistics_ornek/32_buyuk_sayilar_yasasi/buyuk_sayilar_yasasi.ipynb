# 32) Büyük Sayılar Yasası Nedir?
Büyük Sayılar Yasası, bir deney veya işin çok fazla tekrarlandığında elde edilen sonuçların ortalamasının gerçek beklenen değere yaklaşması kuralıdır.

### Temel Özellikler
- Deney sayısı arttıkça sonuçlar gerçeğe daha çok yaklaşır.
- Az sayıda yapılan denemelerde rastgele uç sonuçlar çıkabilir.
- Uzun vadede şans veya risk oranları öngörülebilir hale gelir.

Merkezi Limit Teoremi ile Büyük Sayılar Yasası karıştırılmamalıdır. Merkezi Limit Teoremi bir popülasyonun dağılımı ne olursa olsun bu popülasyondan yeterli büyüklükte (n>=30) çekilen örneklemlerin ortalamalarının normal dağılacağını söylerken, büyük sayılar yasası örneklem boyutu arttıkça örneklem ortalamasının beklenen değere yakınsayacağını söyler.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng()
zar_degerleri = np.array([1,2,3,4,5,6])
print("Büyük Sayılar Yasası Zar Simülasyonu")
print(f"Teorik Beklenen Değer: 3.5\n" + "-" * 45)
for i in [1,10,100,1000,10000,100000,1000000,10000000]:
    zarlar = rng.choice(a=zar_degerleri, size=i, p=[1/6] * 6, replace=True)
    ortalama = np.mean(zarlar)
    print(f'{i} gözlemli için örneklem ortalaması {ortalama}')
```

```python
zarlar = rng.choice(a=zar_degerleri, size=10000, p=[1/6]*6, replace=True)
kumulatif_toplam = np.cumsum(zarlar)
atis_sayilari = np.arange(1, len(zarlar) + 1)
kumulatif_ortalama = kumulatif_toplam / atis_sayilari
plt.figure(figsize=(10,5))
plt.plot(atis_sayilari, kumulatif_ortalama, color='blue')
plt.axhline(y=3.5, color='red', linestyle='--', label='Teorik Beklenen Değer (3.5)')
plt.xscale('log') 
plt.xlabel('Atış Sayısı')
plt.ylabel('Kümülatif Ortalama')
plt.title('Büyük Sayılar Yasası - Zar Simülasyonu')
plt.legend()
plt.show()
```

### Sonuç
Bu örnekte zar atma simülasyonunu 1'den başlatarak, 10000 tane zar atmaya gidene kadar her zar atışında zara gelen sayıların toplamının zar atma adetine bölünmesi ile bulunan attığımız zarlarda çıkan sayıların ortalamasının gittikçe beklenen değere yaklaştığını görüyoruz bir başka deyişle şans faktörü gittikçe azalıyor. Bu da tam Büyük Sayılar Yasası'nın uygulamasıdır.
