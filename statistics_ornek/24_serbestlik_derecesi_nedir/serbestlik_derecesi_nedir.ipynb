# 24) Serbestlik Derecesi (Degree of Freedom) Nedir?
Serbestlik derecesi, bir hesaplamada serbestçe değişebilen değer sayısıdır. n gözlemli bir örneklemde ortalama belliyse geriye kalan n-1 değer serbestçe seçilebilen değerler yani serbestlik derecesidir bu kavramı aşağıda bir örnek üzerinden görürsek daha anlaşılır olacaktır.

Örneğin veri kümemiz 3 gözlemli ve 10 ortalamalı sayılardan oluşsun.
ilk 2 veriyi biz (7,8), (1,2), (3,4) gibi istediğimiz şekilde seçeriz ancak burada 3 sayının toplamı 30 olması gerektiğinden (ki ortalama 10 çıksın) son değerin bizim seçtiğimiz 2 değerin toplamının üstüne otomatik olarak 30'u tamamlayacak şekilde olması gerekir ki serbestlik derecesi n-1'de buradan gelir. Serbestlik derecesi, t dağılımı ve bir çok test istatistiğinde (ANOVA, bağımsız örneklem *t* testi gibi) kullanılmaktadır.
