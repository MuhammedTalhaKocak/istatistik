# 27) Tahmin Teorisi Nedir?
Tahmin Teorisi (Estimation Theory), popülasyonun bilinmeyen bir parametresini (örneğin gerçek ortalama μ veya gerçek standart sapma σ), elimizdeki örneklem verisini kullanarak **tahmin etme** yöntemlerini inceleyen istatistik dalıdır.

Buraya kadar öğrendiğimiz her şey (örneklem ortalaması, standart hata, Z-Score, t-dağılımı) aslında hep bu tek amaca hizmet ediyordu: 
**"Göremediğimiz popülasyon parametresi hakkında, elimizdeki örneklemden yola çıkarak en iyi tahmini nasıl yaparız?"**

### İki Tür Tahmin
Bir popülasyon parametresi iki farklı şekilde tahmin edilebilir:

- **Nokta Tahmini (Point Estimation):** Popülasyon parametresi için **tek bir sayı** vermektir. Örneğin: "Superstore'un tüm işlemlerinde ortalama Profit'in 28.66$ olduğunu tahmin ediyorum." Bu basit ve anlaşılırdır, ama örneklemden kaynaklanan belirsizliği hiç göstermez — sanki kesin bir cevapmış gibi görünür.

- **Aralık Tahmini (Interval Estimation):** Popülasyon parametresinin, belirli bir **güven düzeyiyle** (örneğin %95) hangi **aralıkta** olduğunu söylemektir. Örneğin: "%95 güvenle, gerçek ortalama Profit 25$ ile 32$ arasındadır." Bu, nokta tahmininden çok daha dürüst bir yaklaşımdır çünkü belirsizliği açıkça gösterir.

### Neden Aralık Tahmini Daha Doğru Bir Yaklaşım?
Bir örneklemden hesapladığımız değer (örneğin ortalama), her zaman gerçek popülasyon değerinden bir miktar sapar bunu Standart Hata (Konu 19) konusunda görmüştük. Bu yüzden kesin bir sayı yerine, "gerçek değer 
muhtemelen şu aralıktadır" demek, örneklemden kaynaklanan bu doğal belirsizliği hesaba katan çok daha gerçekçi bir yaklaşımdır.
