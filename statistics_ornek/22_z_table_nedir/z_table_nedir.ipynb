# 22) Z Tablosu Nedir?
Z Tablosu, standart bir normal dağılımda (ortalaması = 0, standart sapması = 1) belirli bir Z skorunun **solunda kalan alanın (kümülatif olasılığın)** ne olduğunu gösteren hazır bir referans tablosudur. Z Tablosundaki değerler önceden hesaplanmış z skorları üzerinden hesaplanan değerleri içerir bu yüzden önce elimizdeki verinin z skorunu hesaplamamız gerekir. 

Z Tablosunun asıl işlevi, Z Skorunu (sadece "ortalamadan kaç standart sapma uzakta" diyen bir sayı) somut bir **olasılık/yüzdeye** çevirmektir. Bu sayede "bu değer ne kadar sıra dışı/olası?" sorusuna sayısal bir cevap verilebilir.

Z Tablosu, ileride göreceğimiz **güven aralığı** hesaplamalarında (örneğin %95 güven aralığı için kullanılan Z=1.96 sabitinin kaynağı) ve **hipotez testinde p-değeri** hesaplamalarında temel bir araç olarak kullanılır.
