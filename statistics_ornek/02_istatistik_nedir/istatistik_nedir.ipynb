# 2) İstatistik Nedir?
İstatistik bilimi, doğru bir şekilde veri toplama, veriyi işleyip anlamlı çıktılara dönüştürme ve veriden geleceğe yönelik tahminler yapmayı içeren bir bilim dalıdır.

## İstatistiğin Temel 2 Dalı
- **Betimsel İstatistik:** Elimizdeki mevcut veriyi (örneklemi) olduğu gibi özetleyip tanımlama sürecidir. Sadece eldeki veri üzerinden bir çıkarım yapılır, popülasyon üzerine bir genelleme yapılmaz.
- **Çıkarımsal İstatistik:** Elimizdeki örneklemden yola çıkarak popülasyon hakkında çeşitli istatistiksel teknikler kullanarak tahmin üretme sürecidir.
