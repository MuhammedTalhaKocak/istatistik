# 33) Olasılık Dağılımı Nedir?
Bir olasılık dağılımı, bir rassal olayın ortaya çıkabilmesi için değerleri ve olasılıkları tanımlar. Örneğin, bir rassal olay olarak madeni paranın tek bir defa havaya atılıp yere düşmesi ele alınsın; değerler 'yazı' veya 'tura' veya bunlar isimsel değişken ölçeğinde ifade edilirse 0 (yazı) veya 1 (tura) olur; olasılıklar ise her iki değer için ½ olacaktır. Böylece madeni bir paranın tek bir defa atılma olayı için iki değer ve ilişkili iki olasılık bu rassal olayın olasılık dağılımı olur. Bir sürekli olasılık dağılımı değerleri bir sürekli olan açıklıkta tanımlar ve tek bir değer için olasılık sıfıra eşittir. Sürekli bir olasılık dağılımında değerler sonsuz olduğu için (örneğin bir kişinin boyu 172.0, 172.01, 172.001... şeklinde sonsuz hassasiyette olabilir), tek bir noktanın olasılığı her zaman sıfıra eşittir çünkü olasılıkların toplamı 1 olmak zorundadır ve bu, sonsuz sayıda değere paylaştırılamaz. Bu yüzden sürekli değişkenlerde "tam olarak şu değer" yerine, her zaman bir **aralığın** olasılığı sorulur (örn: "boyu 170-175 cm arasında olma olasılığı nedir?").
