# One-Hot Dönüşümü (One-Hot Encoding)
One-Hot Encoding, bir kategorik değişkenin **her bir kategorisi için 
ayrı bir 0/1 (ya da True/False) sütunu** oluşturarak, Label Encoding'in 
"yanlış sıralama" tuzağını çözer.

## Nasıl Çalışır?
Bir gözlem, hangi kategoriye aitse, **o kategorinin sütununda 1 (True)**, 
diğer tüm kategori sütunlarında **0 (False)** olur.

**Örnek:** "Kırmızı" olan bir satır:
Kırmızı=1, Mavi=0, Yeşil=0


## Neden Sıralama Sorunu Çözülüyor?
Artık **her kategori kendi bağımsız sütununda** — hiçbir sayısal 
büyüklük ilişkisi (Yeşil>Mavi gibi) yok, her sütun sadece "bu kategori 
mi değil mi" sorusuna cevap veriyor. Bu, nominal ölçeğin (Konu 4) doğal 
yapısına tam uyuyor.

## Python'da Kullanımı
```python
import pandas as pd

onehot = pd.get_dummies(df['Renk'])
print(onehot)
```

## Dikkat — Bir Sonraki Konuyla Bağlantı
One-Hot Encoding uygularken dikkat edilmesi gereken önemli bir tuzak 
var: **Kukla Değişken Tuzağı (Dummy Variable Trap)** — bir sonraki 
konumuz. Kısaca önizleyelim: eğer 3 renk kategorisi varsa ve 3 sütunun 
hepsini tutarsak, bu sütunlar arasında **gereksiz bir bağımlılık** 
oluşur (birini bilirsek diğer ikisini de çıkarabiliriz) — bu, ileride 
regresyon gibi modellerde soruna yol açabilir. Çözüm, genelde bir 
kategoriyi "referans" olarak bırakıp o sütunu çıkarmaktır 
(`pd.get_dummies(drop_first=True)` parametresi).

````python
import pandas as pd

df = pd.DataFrame({'Renk': ['Kırmızı', 'Mavi', 'Yeşil', 'Kırmızı', 'Mavi', 'Yeşil']})

onehot = pd.get_dummies(df['Renk'])
print("One-Hot Encoding (tüm kategoriler):")
print(onehot)
````

### Sonuç
Elimizdeki 3 değerden ikisini bilmemiz üçüncüsünü de bileceğimiz anlamına geldiği için burada 3. bir sütun gereksiz oluyor. Örneğin Kırmızı, Mavi ve Yeşil renkleri varken, 2. satırda kırmızı ve mavi False ise sonuç otomatik olarak Yeşildir. Bu soruna kukla değişken tuzağı denir, sonraki konuda bu sorunu çözmenin yolunu göreceğiz.
