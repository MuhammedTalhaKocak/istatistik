# 15) Veri Seti Tanıyalım

Bir analize başlamadan önce, veri setini sistematik olarak tanımak gerekir aksi halde ilerleyen aşamalarda (eksik değer, aykırı değer, hipotez 
testi) sürprizlerle karşılaşılabilir. Ek olarak, bir veri setini analize hazır hale getirmek için veri setindeki olası sorunları (boş değer, aykırı değer, yanlış veri tipi, normal dağılımın olmaması ...) çözersek ileriki analizlerimizde daha iyi sonuçlar alabiliriz. Bu tanıma süreci genelde şu adımlardan oluşur:

1. **Boyut:** Kaç satır, kaç kolon var? (`.shape`)
2. **Veri Tipleri:** Her kolon hangi tipte kayıtlı? (`.dtypes`) — bu, 
   Konu 3'te gördüğümüz nicel/nitel ayrımını teyit etmemizi sağlar.
3. **Eksik Değer Kontrolü:** Herhangi bir kolonda boş/eksik veri var mı? 
   (`.isnull().sum()`) — çözümünü ileride (Eksik Değer bölümünde) 
   işleyeceğiz, burada sadece "var mı yok mu" tespiti yapıyoruz.
4. **Betimsel Özet:** Sayısal kolonlar için `.describe()` — ortalama, std, 
   min, max, kartiller tek seferde görülür.
5. **Kategorik Kolonların Benzersiz Değerleri:** `.unique()` / `.nunique()` 
   — bir kategorik değişkenin kaç farklı grup içerdiğini gösterir.

Bu 5 adım, bir veri setiyle çalışmaya başlamadan önce atılması gereken 
**ilk keşif (initial exploration)** adımlarıdır.

Örnek olması adına Seaborn kütüphanesinden exercise veri setini inceleyelim.
Ek Bilgi: Pandas, Python'un veriyi analize hazır hale getirme sürecinde en çok kullanılan ve veriyi keşfetmemizi sağlayan açık kaynaklı bir kütüphanedir. Biz de bu kütüphanenin fonksiyonlarından yararlanacağız.

```python
import seaborn as sns
import pandas as pd
df = sns.load_dataset("exercise")
```

```python
# excel'deki tablo yapısı Pandas'ta DataFrame olarak adlandırılır. DataFrame, satır ve sütunlardan oluşan bir veri yapısıdır.
print(f"Veri setinin gözlem (satır) sayısı: {df.shape[0]}")
print(f"Veri setinin sütun sayısı: {df.shape[1]}")
```

```python
print(df.dtypes) # Veri setindeki her bir sütunun veri tipini öğrenmek için dtypes özelliğini kullanabiliriz.
```

```python
# Sütun Bazlı Boş Değer Kontrolü
print(df.isnull().sum())
print()
print(df.isna().sum().sum()) # isna() ve isnull() fonksiyonları aynı işlemi yapar. Her iki fonksiyon da veri setindeki boş değerleri kontrol etmek için kullanılır. 2 kere toplam almak verinin tamamındaki boş hücre sayısını verir.
```

```python
print("Betimsel İstatistikler")
print(df.describe())
```

```python
# Veri setindeki sadece kategorik degiskenleri cekelim.
categorical = df.select_dtypes(include=["category"]) # Belirli bir değişkene ait sütunları filtrelemek için select_dtypes() fonksiyonunu kullanabiliriz. include parametresi ile hangi veri tiplerini seçmek istediğimizi belirtebiliriz. Burada "category" veri tipine sahip sütunları seçiyoruz.
print(categorical)
```

```python
print(df["diet"].unique()) # unique() fonksiyonu, bir sütundaki benzersiz değerleri döndürür.
print(df["diet"].nunique()) # nunique() fonksiyonu, bir sütundaki benzersiz değerlerin sayısını döndürür.
```

```python
print("*****Veri Seti Hakkındaki Genel Bilgiler*****")
df.info() # info() fonksiyonu, veri seti hakkında genel bilgiler verir. Bu bilgiler arasında sütun isimleri, veri tipleri, boş değer sayısı ve bellek kullanımı gibi bilgiler bulunur.
```
