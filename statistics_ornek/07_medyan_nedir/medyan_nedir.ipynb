# 7) Medyan Nedir?
Medyan ya da ortanca, bir anakütle ya da örneklem veri serisinin küçükten büyüğe sıralandığında tam ortada kalan değerdir. Elimizdeki veri seti tek sayıda gözlemden oluşuyorsa bu değer tam ortada kalan değerdir ancak elimizdeki veri seti çift sayıda gözlemden oluşuyorsa medyan değeri tam ortada kalan 2 değerin toplamının ikiye bölünmesiyle hesaplanır.

```python
import numpy as np
tek_degerli_medyan = np.array([5,10,3,20,39,1,9,11,91,10000,37])
print(f"{len(tek_degerli_medyan)} gözlemli veri setinin medyanı: {np.median(tek_degerli_medyan)}")
cift_degerli_medyan = np.array([5,10,3,20,1,9,11,91,19,3000])
print(f"{len(cift_degerli_medyan)} gözlemli veri setinin medyanı: {np.median(cift_degerli_medyan)}") # medyan hesaplamak için NumPy kütüphanesinden median fonksiyonu kullanılır.
```

#### Yukarıda her iki veri seti de 10000 ve 3000 gibi aşırı uç değerler içermesine rağmen medyan sayıları küçükten büyüğe sıralayıp tam ortadaki değeri döndürdüğü için bu değerler veri setini doğru bir şekilde yansıtmaktadır bu durumda medyanın veri setindeki uç değerlerden etkilenmediğini söyleyebiliriz.
