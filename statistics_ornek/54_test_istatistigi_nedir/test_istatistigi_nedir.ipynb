# 54) Test İstatistiği Nedir?
Test İstatistiği, elimizdeki örneklem verisini, **standartlaştırılmış tek bir sayıya** dönüştüren değerdir. Bu sayı, "örneklemden bulduğumuz sonuç, H0'ın iddia ettiği değerden **kaç birim (standart hata cinsinden) uzakta**" sorusuna cevap verir. P-değeri, işte bu tek sayıdan hesaplanır.

## Genel Formül (Neredeyse Her Testte Aynı İskelet)
$$\text{Test İstatistiği} = \frac{\text{Örneklemden Bulunan Değer} - \text{H0'daki Varsayılan Değer}}{\text{Standart Hata}}$$

Bu formül, Konu 20'de (Z-Score) gördüğümüz $Z = \frac{x-\bar{x}}{s}$ formülünün, çıkarımsal istatistiğe uyarlanmış hali. Sadece pay ve payda biraz değişiyor: tek bir gözlem yerine örneklem ortalaması, std yerine standart hata kullanılıyor.

## Hangi Test İstatistiği, Hangi Duruma?
Hangi formülün/dağılımın kullanılacağı, testin türüne göre değişir:
- **Z-istatistiği:** Popülasyon std'si biliniyorsa veya n≥30
- **t-istatistiği:** Popülasyon std'si bilinmiyorsa ve n<30
- İlerleyen konularda (Ki-Kare, F-istatistiği gibi) farklı test türleri için farklı test istatistikleri de göreceğiz

## Test İstatistiğinden P-Değerine Nasıl Gidilir?
1. Formülle test istatistiğini (örn: Z=2.3 gibi bir sayı) hesaplarız.
2. Bu sayıyı, ilgili dağılıma (Z-Table veya t-Table) göre bir **olasılığa** çeviririz, işte bu olasılık, p-değeridir.
3. P-değerini α ile karşılaştırıp karar veririz (Konu 50).

## Özetle Zincir
$$\text{Veri} \to \text{Test İstatistiği (Z veya t)} \to \text{P-değeri} \to \text{Karar (H0'ı reddet/reddedemedik)}$$

Bu zincir, bundan sonraki tüm hipotez testi konularının (Tek Örneklem Z Testi, T Testi, ANOVA, vs.) ortak iskeleti olacak. Sadece "Test İstatistiği" adımındaki formül, teste göre değişecek.
