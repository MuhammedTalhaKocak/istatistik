# Doğrusallık (Linearity) Nedir?
Doğrusallık, iki değişken arasındaki ilişkinin, bir **düz çizgiyle** 
iyi temsil edilip edilemeyeceğini ifade eder. Bu kavram kritik çünkü 
**Pearson Korelasyonu SADECE doğrusal ilişkileri ölçer** — 
güçlü ama doğrusal olmayan bir ilişki varsa, Pearson bunu **tamamen kaçırabilir**, yanıltıcı şekilde düşük bir r değeri verebilir.

## Neden Bu Kadar Önemli — Çarpıcı Bir Örnek
İki değişken arasında **mükemmel bir ilişki** olsun, ama bu ilişki bir 
**parabol (U şekli)** olsun (örn: $y = x^2$). Bu durumda:
- Değişkenler arasında **tam ve kusursuz bir matematiksel ilişki var** 
- Ama bu ilişki **doğrusal değil**
- Pearson korelasyonu hesaplarsak, sonuç **r≈0** çıkabilir — sanki hiç ilişki yokmuş gibi!

Bu, korelasyonun en tehlikeli tuzaklarından biri: **"Korelasyon düşük, 
demek ki ilişki yok" sonucuna varmak YANLIŞ olabilir** — ilişki güçlü 
olabilir, sadece doğrusal değildir.

## Doğrusallığı Nasıl Kontrol Ederiz?
En basit ve en güvenilir yöntem: **Scatter Plot (Saçılım Grafiği)** 
çizip **gözle** bakmak. Sayısal bir test de var (Ramsey RESET Test gibi, 
ismini bilmemiz yeterli, ileri seviye) ama pratikte scatter plot yeterli 
bir ilk kontrol sağlar.

## Ne Yapılır Doğrusal Değilse?
1. **Veriyi dönüştürmek** (Konu 13'teki log/karekök gibi dönüşümler, bazen doğrusal olmayan ilişkiyi doğrusala çevirebilir)
2. **Spearman Korelasyonu kullanmak** (Konu 91 — Pearson'dan farklı olarak, sadece "monoton" ilişkileri yakalar, tam doğrusallık gerektirmez, daha esnektir)
3. Doğrusal olmayan modelleme teknikleri kullanmak.

## Regresyon İçin Neden Kritik?
İleride göreceğimiz Doğrusal Regresyon'un (Konu 115+) temel varsayımlarından biri de doğrusallıktır — bu konuyu şimdi sağlam anlamamız, oraya geldiğinde büyük fayda sağlayacaktır.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

np.random.seed(42)

# U şeklinde (parabolik) mükemmel bir ilişki kuruyoruz: y = x^2
x = np.linspace(-10, 10, 100)
y = x**2 + np.random.normal(0, 2, 100)  # az gürültü ekliyoruz

pearson_r, pearson_p = stats.pearsonr(x, y)
print(f"Pearson r: {pearson_r:.4f}, p-değeri: {pearson_p:.4f}")

plt.scatter(x, y, alpha=0.6)
plt.title(f"Y = X² İlişkisi (Pearson r = {pearson_r:.4f})")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

Yukarıda X ve Y değişkenleri arasında doğrusal bir ilişki olmasa da U 
şeklinde mükemmel bir parabolik ilişki bulunmaktadır. Ancak Pearson 
korelasyonu bu ilişkiyi temsil edememektedir; Pearson korelasyon 
katsayısına göre bu iki değişken arasında neredeyse hiç ilişki yoktur 
(r=0.0026). Bu iki değişkenin ilişkisi istatistiksel olarak da anlamsız 
bulunmuştur (p=0.979>0.05). Bu durum, korelasyon katsayısının yalnızca 
doğrusal ilişkileri yakalayabildiğini, doğrusal olmayan güçlü 
ilişkilerde yanıltıcı sonuçlar verebileceğini göstermektedir.
