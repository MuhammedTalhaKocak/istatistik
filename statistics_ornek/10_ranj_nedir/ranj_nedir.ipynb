# 10) Ranj Nedir?
Ranj (Range), bir veri kümesindeki en büyük değer ile en küçük değer arasındaki farktır.
- **Ranj = Maximum Değer - Minimum Değer**

- Ranj hesaplanması kolay olmasına karşın, sadece iki uç noktaya **min-max** baktığından dolayı, aradaki diğer tüm gözlemleri göz ardı eder. Bu durumda veri setindeki tek bir aykırı değer, geri kalan veri grubu birbirine çok yakın olsa dahi, Ranj'ı olduğundan çok daha büyük gösterebilir.
- Bu zayıflık nedeniyle Ranj tek başına güvenilir bir yayılım ölçüsü değildir genellikle IQR (Çeyrekler Arası Açıklık) gibi uç değerlerden daha az etkilenen ölçülerle birlikte değerlendirilir.

```python
import numpy as np
ornek1 = np.array([1,2,3,4,5,6,7,100]) # uç değer koyup Ranj'ın nasıl değiştiğini gözlemleyelim.
ornek2 = np.array([1,2,3,4,5,6,7,8])   # Bu örnekte de ilk 7 veriyi yukarıdaki örnekle aynı tuttuk ve herhangi bir uç değer eklemedik.
print(f"Örnek 1'in Ranjı: {np.max(ornek1) - np.min(ornek1)}") 
print(f"Örnek 2'in Ranjı: {np.max(ornek2) - np.min(ornek2)}") 
```

Yukarıdaki iki örnekte her iki veri kümesinin de ilk 7 değeri aynı olmasına rağmen ilk örneğin uç değer bulundurmasından ötürü Ranj'ın alttaki örneğe kıyasla yaklaşık 14 kat daha büyük çıktığını gözlemleyebiliriz.
