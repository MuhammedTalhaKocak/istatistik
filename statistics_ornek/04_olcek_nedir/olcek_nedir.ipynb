# 4) Ölçek Nedir?
Ölçek, nesnelerin veya olayların özelliklerine sayı veya sembol vermek için kullanılan kural ve yöntemler bütünüdür. Verilerin türünü ve hangi matematiksel işlemlerin yapılabileceğini belirler.
## Ölçek Türleri
- **Sınıflama (Nominal)**: Aralarında bir üstünlük derecesi bulunmayan bir grup içindeki farklı değerleri sayılar ile ifade etme sürecidir. Örneğin, Cinsiyet özelinde Kadınlara 1, Erkeklere 0 değerinin atanması. Buradaki 1 ve 0 sayıları sadece birer etikettir, aralarında büyüklük-küçüklük ilişkisi yoktur — sadece iki farklı grubu ayırt etmek için kullanılır.
- **Sıralama (Ordinal)**: Bir gruptaki farklı değerler arasında hiyerarşi bulunduğu durumda kullanılan ölçek türüdür. Bu değerler arasında bir fark bulunsa da bu fark'ın ne kadar büyüklükte olduğuna dair kesin yorumlar yapılamaz.
- **Aralık (Interval)**: Eşit aralıklara sahip ve 0 noktasının yokluk anlamını içermediği değişken türüdür. Örneğin, Sıcaklık (Sıcaklığın 0 dereceye eşit olması sıcaklığın olmadığı değil havanın soğuk olduğu anlamına gelir.)
- **Oran (Ratio)**: Bu ölçek türünde interval ölçek türünden farklı olarak 0 noktası yokluğu ifade eder. Örneğin, İndirim (İndirimin %0'a eşit olması herhangi bir indirim uygulanmadığını ifade eder.)
