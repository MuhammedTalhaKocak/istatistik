# Eksik Değer (Missing Value) Nedir?
Eksik Değer, bir veri setinde bazı gözlemler için **bazı değişkenlerin 
kaydedilmemiş/bilinmeyen** olması durumudur (pandas'ta genelde `NaN` 
olarak görünür). Bu, Veri Önişleme sürecinin en kritik adımlarından 
biridir `isnull().sum()` ile "Eksik gözlem var mı yok mu" diye bakmıştık, şimdi bunu **derinlemesine** ele alıyoruz.

## Neden Önemli?
Çoğu istatistiksel test ve ML algoritması, eksik değerlerle **doğrudan 
çalışamaz** — ya hata verir ya da o satırı otomatik (ve bazen 
sessizce/tehlikeli şekilde) atar. Eksik değerleri nasıl ele aldığın, 
analizinin sonucunu ciddi şekilde etkileyebilir.

## Eksik Değerin Kaynakları
- Veri girişi hataları (biri bir alanı doldurmayı unutmuş)
- Sensör/sistem arızaları
- Katılımcının cevap vermeyi reddetmesi (anketlerde çok yaygın)
- Veri birleştirme (join/merge) işlemlerinde eşleşme bulunamaması

## Eksik Değer Türleri 
Eksik değerlerin **neden** eksik olduğu, onları nasıl ele almamız
gerektiğini belirler. Üç ana tür vardır:
- **MCAR (Missing Completely At Random):** Eksiklik tamamen rastgele, 
  hiçbir değişkenle ilişkili değil
- **MAR (Missing At Random):** Eksiklik, **başka bir gözlenen 
  değişkenle** ilişkili (ama eksik değerin kendisiyle değil)
- **MNAR (Missing Not At Random):** Eksiklik, **değerin kendisiyle** 
  ilişkili (en tehlikeli tür)

## Python'da Temel Tespit
```python
df.isnull().sum()           # Her kolonda kaç eksik değer var
df.isnull().sum() / len(df) # Yüzde olarak
df.isnull().any(axis=1)     # Hangi satırlarda en az bir eksik var
```

## Görselleştirme
`missingno` kütüphanesi, eksik değer örüntülerini görsel olarak 
incelemek için kullanılır (senin dev ortamında zaten kurulu):
```python
import missingno as msno
msno.matrix(df)
```
