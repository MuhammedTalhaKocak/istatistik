# Spearman Korelasyon Testi
Spearman, Pearson'ın (Konu 90) iki büyük zayıflığına (doğrusallık 
gereksinimi, aykırı değer hassasiyeti) karşı **dayanıklı (robust)** bir 
alternatiftir — Non-Parametrik Testler bölümünde gördüğümüz Mann-Whitney/
Kruskal-Wallis'in kullandığı **"sıralama (rank)" mantığının**, 
korelasyona uygulanmış hali.

## Nasıl Çalışır?
Ham değerler yerine, her değişkenin **sıralamasını (rank)** alır, sonra 
Pearson formülünü bu sıralamalara uygular. Yani "X arttıkça Y de artıyor 
mu" yerine, "**X'in sırası** arttıkça **Y'nin sırası** da artıyor mu" 
sorusunu sorar.

## Neden Daha Dayanıklı?
- **Aykırı değerlere karşı:** 200 gibi bir aşırı değer, sıralamada 
  sadece "en büyük" (yani son sıra) olur — büyüklüğü ne kadar aşırı 
  olursa olsun, sıralamadaki etkisi sınırlı kalır (bir önceki konuda 
  gördüğümüz "tek noktanın domine etmesi" sorunu büyük ölçüde ortadan 
  kalkar)
- **Doğrusallık gerektirmez:** Sadece **monoton** bir ilişki yeterli 
  (yani X arttıkça Y'nin sürekli artması VEYA sürekli azalması yeterli 
  — düz bir çizgi olması şart değil, eğrisel de olabilir, yeter ki yön 
  tutarlı olsun)

## Monoton İlişki Nedir? (Pearson'ın Kaçırdığı U-Şekli ile Farkı)
Dikkat: Spearman da **U-şekli gibi yön değiştiren** (önce azalıp sonra 
artan) ilişkileri yakalayamaz — çünkü o da "yön" arıyor. Ama **eğrisel 
olsa bile TEK YÖNDE giden** ilişkilerde (örn: $y=x^3$ gibi, sürekli 
artan ama doğrusal olmayan) Pearson'dan çok daha başarılıdır.

## Hipotezler
- **H0:** İki değişken arasında monoton bir ilişki yoktur
- **H1:** İki değişken arasında monoton bir ilişki vardır

## Python'da Kullanımı
```python
from scipy.stats import spearmanr

rho, p_degeri = spearmanr(x, y)
```
(Spearman'ın korelasyon katsayısına genelde $\rho$ - "rho" denir, $r$ 
değil — ama yorumlama mantığı birebir aynı: -1 ile +1 arası, işaret ve 
büyüklük aynı şekilde okunur.)

## Ne Zaman Pearson, Ne Zaman Spearman?
- Veri **doğrusal + normal + aykırı değersiz** ise → Pearson (daha 
  güçlü/hassas)
- Veri **sıralı (ordinal)**, **doğrusal değil ama monoton**, veya 
  **aykırı değer içeriyorsa** → Spearman (daha güvenli)

````python
from scipy.stats import spearmanr, pearsonr
import numpy as np
x = np.random.normal(50, 10, 30)
y = np.random.normal(50, 10, 30)

# --- Aykırı Değer Yokken ---
r_once, p_once = pearsonr(x, y)
print(f"Pearson (aykırı değer YOK): r={r_once:.4f}, p={p_once:.4f}")
rho_once, rho_p_once = spearmanr(x, y)
print(f'Spearman (aykırı değer YOK): rho:{rho_once:.4f}, p={rho_p_once:.4f}')

# --- Aykırı Değer Varken ---
x_aykiri = np.append(x, 200)
y_aykiri = np.append(y, 200)
r_sonra, p_sonra = pearsonr(x_aykiri, y_aykiri)
print(f"Pearson (aykırı değer VAR): r={r_sonra:.4f}, p={p_sonra:.4f}")
rho_sonra, rho_p_sonra = spearmanr(x_aykiri, y_aykiri)
print(f'Spearman (aykırı değer VAR): rho={rho_sonra:.4f}, p={rho_p_sonra:.4f}')

print('Karşılaştırma:')
print(f'Pearson - Aykırı değer yok: r={r_once} | Aykırı değer var: r={r_sonra}')
print(f'Spearman - Aykırı değer yok: rho={rho_once} | Aykırı değer var: rho={rho_sonra}')
````

### Sonuç
Pearson korelasyonu aykırı değer yokken ilişkinin negatif olduğunu ve anlamsız olduğunu söylerken 1 aykırı değer ile ilişki hem pozitif hem de anlamlı bir ilişkiye dönüştü bu pearson korelasyonun aykırı değerlerden ne kadar çok etkilendiğinin bir kanıtı. Öte yandan, spearman korelasyonunda ise aykırı değer yokken ve varken her iki durumda da ilişki hem negatif hem de anlamsız çıktı. Bu sonuç da, spearman korelasyonunun aykırı değerlere karşı dirençli (robust) olduğunun kanıtıdır.
