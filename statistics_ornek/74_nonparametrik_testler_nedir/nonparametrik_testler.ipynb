# Non-Parametrik Testler Nedir?
Şimdiye kadar öğrendiğimiz testlerin (Z, T, ANOVA, MANOVA) hepsi **parametrik** testlerdi yani belirli varsayımlara (özellikle **normallik**) dayanıyorlardı. Non-Parametrik testler, bu varsayımlara **ihtiyaç duymadan** çalışan alternatif testlerdir.

## Ne Zaman Kullanılır?
1. **Normallik sağlanmadığında:** Shapiro-Wilk testi H0'ı reddettiğimizde (veri normal değilse) ve veri dönüşümü (log, karekök gibi) de işe yaramadığında
2. **Örneklem çok küçük olduğunda:** Normalliği güvenilir test edemeyecek kadar az veri varsa (n<10 gibi)
3. **Veri sıralı (ordinal) olduğunda:** Değerler sayısal değil, sadece sıralama bilgisi taşıyorsa (Konu 4'teki ölçek türleri örn: "Çok Kötü/Kötü/Orta/İyi/Çok İyi" gibi bir memnuniyet anketi)

## Parametrik Testlerin Non-Parametrik Karşılıkları
Her parametrik testin, genelde bir non-parametrik "kardeşi" vardır:

| Parametrik Test | Non-Parametrik Karşılığı |
|---|---|
| Tek Örneklem T Testi | İşaret Testi (Sign Test) |
| Bağımsız İki Örneklem T Testi | Mann-Whitney U Testi |
| Bağımlı Örneklem T Testi | Wilcoxon İşaretli Sıralar Testi |
| One-Way ANOVA | Kruskal-Wallis H Testi |
| Repeated Measures ANOVA | Friedman Testi |

## Nasıl Çalışırlar? (Genel Mantık)
Non-parametrik testler, ham veri değerleri yerine genelde verinin **sıralamasını (rank)** kullanır yani "değerlerin ne kadar büyük 
olduğu" yerine "hangi sırada olduğu" ile çalışır. Bu, uç değerlerden (outlier) ve dağılım şeklinden **daha az etkilenmelerini** sağlar 
— parametrik testlere göre daha "sağlam/dayanıklı" (robust) olurlar.

## Dezavantajı
Genelde parametrik testlerden **daha az istatistiksel güce (Power) sahiptirler** yani varsayımlar gerçekten sağlanıyorsa, parametrik 
test tercih edilmelidir (gerçek etkileri daha iyi yakalar). Non-parametrik testler, sadece varsayımlar sağlanmadığında bir "güvenlik ağı" olarak kullanılmalıdır.
