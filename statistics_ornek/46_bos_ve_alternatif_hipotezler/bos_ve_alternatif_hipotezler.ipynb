# 46) Boş ve Alternatif Hipotezler
Her hipotez testi, birbirine **zıt** iki iddiadan oluşur:

## Boş Hipotez (H₀ - Null Hypothesis)
**"Hiçbir fark/etki yok"** iddiasıdır. Mevcut durumun, statükonun devam ettiğini varsayar. Konu 48'deki örnekte "para adildir" H₀ idi.

Hipotez testi her zaman **H₀'ın doğru olduğunu varsayarak** başlar (tıpkı bir mahkemede "suçlu kanıtlanana kadar masumdur" mantığı gibi) H₀'ı çürütmek için güçlü kanıt aranır.

## Alternatif Hipotez (H₁ veya Hₐ - Alternative Hypothesis)
**"Bir fark/etki VAR"** iddiasıdır, H₀'ın tam tersi. Bu, genelde araştırmacının **kanıtlamaya çalıştığı** iddiadır. Konu 48'deki örnekte 
"para hileli" H₁ idi.

## Örnekler (Senin Alanından)
| Senaryo | H₀ (Fark Yok) | H₁ (Fark Var) |
|---|---|---|
| Yeni buton rengi | Tıklama oranı değişmedi | Tıklama oranı değişti |
| Pazarlama kampanyası | Satışlara etkisi yok | Satışları artırdı/azalttı |
| İki mağaza karşılaştırması | İki mağazanın ortalama satışı aynı | Mağazalar arasında fark var |

## Test Sonrası İki Olası Karar
Hipotez testinin sonunda sadece iki karar verebiliriz:
1. **H₀'ı reddet** → Veriler, H₀'ın doğru olma ihtimalinin çok düşük olduğunu gösteriyor, demek ki H₁ (fark var) daha makul
2. **H₀'ı reddedemedik** → Veriler, H₀'ı çürütecek kadar güçlü bir kanıt sunmuyor (dikkat: bu, "H₀ kesinlikle doğrudur" demek DEĞİLDİR sadece "onu reddetmek için yeterli kanıt yok" demektir)

## Önemli İnce Nokta
"H₀'ı reddedemedik" ile "H₀ doğrudur" **aynı şey değildir.** Tıpkı mahkemede "beraat etti" demenin "kesinlikle masum" demek olmadığı gibi sadece "suçlu olduğuna dair yeterli kanıt bulunamadı" demektir. Bu ayrım, hipotez testi sonuçlarını doğru yorumlamak için kritiktir.
