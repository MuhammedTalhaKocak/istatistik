# 82) Korelasyon Nedir?
Korelasyon, iki **sürekli değişken** arasındaki **doğrusal ilişkinin 
yönünü ve gücünü** ölçen istatistiksel bir kavramdır. Konu (Varyans-
Kovaryans Matrisi)'nda gördüğümüz kovaryansın, **standartlaştırılmış** 
halidir. Hatırlarsak orada "ham kovaryans sayılarının büyüklüğü 
yanıltıcı olabilir, standartlaştırmamız gerekir" demiştik, işte bu 
standartlaştırma korelasyondur.

## Kovaryanstan Farkı
$$r = \frac{Cov(X,Y)}{s_X \cdot s_Y}$$

Kovaryansı, iki değişkenin **kendi standart sapmalarına** bölerek, 
sonucu her zaman **-1 ile +1 arasına** sıkıştırıyoruz. Bu sayede farklı 
ölçeklerdeki (örn: TL cinsinden Satış ile adet cinsinden Miktar) 
ilişkileri **karşılaştırılabilir** hale getiriyoruz — ham kovaryans 
bunu yapamıyordu (Varyans-Kovaryans örneğimizdeki 6894 gibi büyük 
sayılar, büyüklük olarak yorumlanamıyordu).

## Korelasyon Katsayısının (r) Yorumlanması
- **r = +1:** Mükemmel pozitif doğrusal ilişki (biri artarken diğeri de tam orantılı artar)
- **r = -1:** Mükemmel negatif doğrusal ilişki (biri artarken diğeri tam orantılı azalır)
- **r = 0:** Doğrusal ilişki yok
- **Kaba yorumlama eşikleri** (Cohen'in önerdiği, kesin kurallar değil):
  - |r| < 0.3 → Zayıf
  - 0.3 ≤ |r| < 0.7 → Orta
  - |r| ≥ 0.7 → Güçlü

## En Kritik Uyarı — Korelasyon Nedensellik Değildir!
İki değişken arasında güçlü bir korelasyon bulmak, birinin **diğerine 
sebep olduğu** anlamına gelmez. Klasik örnek: "Dondurma satışları ile 
boğulma vakaları arasında güçlü pozitif korelasyon var" — ama dondurma 
boğulmaya sebep olmuyor, ikisi de **sıcak hava** (gizli/karıştırıcı 
değişken, tıpkı ANCOVA'da gördüğümüz Kanal-Yaş-Memnuniyet örneği gibi) 
tarafından tetikleniyor.

## Sadece Doğrusal İlişkiyi Yakalar
Korelasyon (en azından Pearson, Konu 90) **sadece doğrusal (linear)** 
ilişkileri ölçer. İki değişken arasında çok güçlü ama **doğrusal 
olmayan** (örn: U şeklinde) bir ilişki olsa bile, korelasyon katsayısı 
sıfıra yakın çıkabilir — bu, Konu 89'da (Doğrusallık) daha detaylı 
göreceğimiz önemli bir tuzak.

## Neden Önemli?
Regresyon analizinin (Konu 113+) temelini oluşturur — regresyon, aslında 
korelasyonun "bir değişkenden diğerini tahmin etmeye" genişletilmiş 
halidir.
