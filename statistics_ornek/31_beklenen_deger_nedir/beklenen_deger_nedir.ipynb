# 31) Beklenen Değer Nedir?
Beklenen Değer $E(X)$, bir rassal değişkenin **uzun vadede, deney çok sayıda tekrarlandığında ortalama olarak alacağı değerdir.** Tek bir 
denemede bu değeri hiç görmeyebiliriz. Beklenen Değer, "gerçekleşen" bir sonuç değil, **teorik/uzun vadeli bir ortalamadır.**

Rassal değişkenin kesikli mi sürekli mi olduğuna göre Beklenen değer hesaplama yöntemi değişir:

### Kesikli Rassal Değişken İçin
Olası her değeri, o değerin olasılığıyla çarpıp toplarız:
$$E(X) = \sum_{i} x_i \cdot P(x_i)$$

*Örnek Zar Atışı:* Zar 6 farklı değer alabilir (1-6), her birinin 
olasılığı $\frac{1}{6}$'dır:
$$E(X) = 1\cdot\tfrac{1}{6} + 2\cdot\tfrac{1}{6} + 3\cdot\tfrac{1}{6} + 4\cdot\tfrac{1}{6} + 5\cdot\tfrac{1}{6} + 6\cdot\tfrac{1}{6} = 3.5$$

Dikkat: **3.5, zarın hiçbir yüzünde yazmıyor!** Bu, Beklenen Değer'in gerçekte gerçekleşen bir sonuç olmak zorunda olmadığını, sadece çok sayıda atışın **ortalamasının** yakınsayacağı teorik bir nokta olduğunu gösterir.

### Sürekli Rassal Değişken İçin
Sürekli değişkenlerde olası değerler sonsuz olduğu için, toplama (∑) yerine **integral** kullanılır:
$$E(X) = \int_{-\infty}^{\infty} x \cdot f(x)\, dx$$

Burada $f(x)$, değişkenin **olasılık yoğunluk fonksiyonudur**. Mantık aynıdır sadece kesikli toplama yerine sürekli bir "toplama" (integral) işlemi yapılır çünkü değerler artık sayılabilir/ayrık değil, kesintisiz bir aralıkta yayılmıştır.

### Ortak Nokta
İki formül de aynı temel fikri taşır: **her olası değeri, o değerin ne kadar olası olduğuyla ağırlıklandırıp ortalamasını almak.** Kesiklide bunu tek tek toplayarak, süreklide integral alarak yaparız matematiksel araç farklı ama amaç birebir aynıdır.
