# 58) Bağımsız İki Örneklem T Testi
Bu teste kadar hep **bir örneklem ortalamasını, sabit bir değerle** karşılaştırıyorduk (Z ve T testlerinde). Şimdi ilk kez **iki farklı 
grubun ortalamalarını birbirleriyle** karşılaştıracağız. Konu 29'da (İki Popülasyon Ortalama Farkı Güven Aralığı) tanışmıştık, şimdi bunu 
resmi bir hipotez testi çerçevesinde ele alıyoruz.

## Ne Zaman Kullanılır?
İki **bağımsız** (birbirinden etkilenmeyen, farklı bireylerden oluşan) grup arasında ortalama farkı olup olmadığını test etmek istediğimizde. Örnek: "A grubu" ve "B grubu" farklı kişilerden oluşuyorsa (aynı kişiler iki kere ölçülmüyorsa) bağımsızdır.

## Hipotezler
- **H0:** $\mu_1 = \mu_2$ (iki grubun popülasyon ortalamaları eşittir, fark yoktur)
- **H1:** $\mu_1 \ne \mu_2$ (çift kuyruklu) veya $\mu_1 > \mu_2$ / $\mu_1 < \mu_2$ (tek kuyruklu)

## Test İstatistiği Formülü
$$t = \frac{(\bar{x}_1 - \bar{x}_2) - 0}{SE_{fark}}$$

Bu formül, Konu 31'de güven aralığı için kullandığımız yapının aynısı sadece burada "0'dan farkı" test ediyoruz (H0'ın iddiası zaten"fark sıfırdır" olduğu için).

## İki Alt Versiyon (Konu 31'i Hatırla)
1. **Varyanslar eşit varsayılırsa (Pooled):** $SE_{fark}$, iki grubun varyansları "havuzlanarak" hesaplanır, $df = n_1+n_2-2$
2. **Varyanslar eşit varsayılmazsa (Welch):** $SE_{fark}$ aynı formülle ama df, Welch-Satterthwaite formülüyle hesaplanır.

**Hangisini kullanacağımızı nasıl belirleriz?** Varyansların eşit olup olmadığını önce test etmemiz gerekir bu, bir sonraki konumuz olan 
**Varyansların Homojenliği** testiyle belirlenir.

## Python'da Kullanımı
`scipy.stats.ttest_ind()` fonksiyonu, `equal_var` parametresiyle iki versiyonu da destekler:
- `equal_var=True` → Pooled yöntem
- `equal_var=False` → Welch yöntemi

```python
import numpy as np 
from scipy import stats
# Senaryo: İki Farklı Reklam Kreatifinin Sepet Tutarına Etkisi
# Bir e-ticaret şirketi, iki farklı reklam kreatifi (A ve B) kullanarak müşteri çekiyor. Her kreatiften gelen müşterilerin ortalama sepet tutarını (TL) karşılaştırmak istiyorlar hangi kreatif daha "değerli" müşteri getiriyor?

kreatif_A = np.array([320, 285, 410, 295, 350, 275, 390, 310, 265, 335, 300, 355])
kreatif_B = np.array([245, 260, 230, 275, 250, 210, 265, 240, 255, 220, 235, 248])

# H0: kreatif_A ile kreatif_B arasında müşterilerin ortalama sepet tutarı bakımından fark yoktur.
# H1: kreatif_A ile kreatif_B arasında müşterilerin ortalama sepet tutarı bakımından fark vardır.
ttest = stats.ttest_ind(a=kreatif_A, b=kreatif_B, alternative='two-sided') # varyansların homojenliğini henüz öğrenmediğimiz için equal_var parametresini varsayılan olarak bırakıyoruz.
# ttest objesi t istatistiği, p değeri, serbestlik derecesi ve güven aralığı olmak üzere 4 özelliğe sahip.
alpha = 0.05
if (p_value:=ttest.pvalue) < alpha:
    print('H0 hipotezini reddediyoruz, Kreatif A ile Kreatif B"nin ortalamaları arasında anlamlı bir fark var.')
else:
    print('Elimizde H0 hipotezini reddecek kadar güçlü bir kanıt yok, bu durumda Kreatif A ile Kreatif B"nin ortalamaları arasında anlamlı bir fark olmadığı söylenebilir.')
print(f"P değeri: {round(p_value,4)}")
print(f"T istatistiği: {ttest.statistic}")
ci = ttest.confidence_interval(confidence_level=0.95)
print(f"Güven aralığı alt sınır: {ci.low}, Güven aralığı üst sınır: {ci.high}") # ttest objesi üzerinden güven aralığı fonksiyonu ile istediğimiz alfa değeri üzerinden güven aralığı hesaplayıp sonucu ci isimli güven aralığı nesnesine attık. Bu nesne üzerinden de low ve high niteliklerini kullanarak alt ve ust sınırı yazdırıyoruz.
print(f"Serbestlik Derecesi: {ttest.df}")
print(f'Kreatif A ortalama: {np.mean(kreatif_A)}')
print(f'Kreatif B ortalama: {np.mean(kreatif_B)}')
print(f'Kreatif A Standart Sapma: {np.std(kreatif_A)}')
print(f'Kreatif B Standart Sapma: {np.std(kreatif_B)}')
print(f'Kreatif A ortalama - Kreatif B ortalama = {np.mean(kreatif_A) - np.mean(kreatif_B)}')
```

### Sonuç
Yapılan bağımsız örneklem *t*-testi sonuçlarına göre Kreatif A ile Kreatif B arasında müşterilerin ortalama sepet tutarları bakımından anlamlı bir fark vardır (p<0.001***). Ek olarak Kreatif A'daki müşterilerin ortalama sepet tutarları Kreatif B'deki müşterilerin ortalama sepet tutarlarından yaklaşık olarak 80 TL fazladır. Son olarak Kreatif A'daki müşterilerin sepet tutarlarının dağılımı daha geniş iken (Kreatif A standart sapma > Kreatif B standart sapma), Kreatif B'deki müşterilerin sepet tutarlarının dağılımı daha dar bir aralıktadır.
