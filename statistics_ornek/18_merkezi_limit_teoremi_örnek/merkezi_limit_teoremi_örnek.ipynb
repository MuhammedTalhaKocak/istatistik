```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import skew
# 1. Parametrelerimizi Tanımlayalım.
orneklem_boyutu = 30    # Her bir örneklemdeki gözlem sayısı (n >= 30 önerilir)
orneklem_sayisi = 1000  # Popülasyondan 1000 tane örneklem çekeceğiz.

# 2. Normal dağılıma uymayan bir ana kütle oluşturalım. Örnek olarak üstel dağılım kullanabiliriz.
ana_kutle = np.random.exponential(scale=2, size=100000)

# 3. Örneklem Ortalamalarını bu boş listeye ekleyeceğiz.
orneklem_ortalamalari = []

for _ in range(orneklem_sayisi):
    # Ana kütleden rastgele n adet veri seçiyoruz
    orneklem = np.random.choice(ana_kutle, size=orneklem_boyutu)
    # Bu n adet verinin ortalamasını alıp listeye ekliyoruz
    orneklem_ortalamalari.append(np.mean(orneklem))
# Çizim alanını oluşturalım (1 satır, 2 sütun boyutunda)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Sol Grafik: Orijinal Ana Kütle Dağılımı (Üstel / Sağa Çarpık)
ax1 = axes[0]
sns.histplot(ana_kutle, kde=True, ax=ax1, color='red')
ax1.set_title(f"Orijinal Ana Kütle Dağılımı (Çarpıklık:{round(skew(a=ana_kutle),4)})\n(Normal Değil)")
ax1.set_xlabel("Değerler")

# Sağ Grafik: Örneklem Ortalamalarının Dağılımı (Normal Dağılım / Çan Eğrisi)
ax2 = axes[1]
sns.histplot(orneklem_ortalamalari, kde=True, ax=ax2, color='blue')
ax2.set_title(f"Örneklem Ortalamalarının Dağılımı (Çarpıklık:{round(skew(a=orneklem_ortalamalari),4)}\n(n = {orneklem_boyutu}, Simülasyon = {orneklem_sayisi})")
ax2.set_xlabel("Örneklem Ortalaması")

plt.tight_layout() # görsellerin birbirine girmemesi için tight_layout() fonksiyonunu kullanıyoruz.
plt.show() # görsellerin gösterilmesi için plt.show() fonksiyonunu kullanıyoruz.
```

Sonuç olarak, Merkezi Limit Teoremi sayesinde popülasyon normal dağılmadığı halde bu popülasyondan çekilen 1000 tane 30 gözlemli örneklemlerin ortalamasının normal dağıldığını görüyoruz.Ek olarak popülasyon çarpıklığı ve örneklem ortalamalarının çarpıklığı da verilerin nasıl bir dağılıma sahip olduğunu göstermektedir.
