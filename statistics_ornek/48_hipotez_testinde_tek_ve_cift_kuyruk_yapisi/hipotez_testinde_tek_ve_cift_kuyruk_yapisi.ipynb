# 48) Hipotez Testinde Tek ve Çift Kuyruk Yapısı
Bir hipotez testi kurarken, Alternatif Hipotezin (H₁) **yönünü** de belirlememiz gerekir bu, testin "tek kuyruklu" mu "çift kuyruklu" mu 
olacağını belirler.

## Çift Kuyruklu Test (Two-Tailed)
H₁, sadece **"bir fark var"** der, farkın yönü (artma mı azalma mı) önemli değildir. İki uç de (hem çok yüksek hem çok düşük) "anlamlı" sayılır.

**Örnek:**
- H₀: Yeni tasarımın dönüşüm oranı, eskisiyle **aynıdır**
- H₁: Yeni tasarımın dönüşüm oranı, eskisinden **farklıdır** (artmış da 
  olabilir, azalmış da)

Bu durumda α (0.05), dağılımın **iki ucuna paylaştırılır**. Her kuyrukta α/2 = 0.025 olur.

## Tek Kuyruklu Test (One-Tailed)
H₁, **belirli bir yönü** iddia eder. Sadece artış ya da sadece azalış.

**Örnek:**
- H₀: Yeni pazarlama kampanyası, satışları **artırmamıştır** (aynı ya da azalmıştır)
- H₁: Yeni pazarlama kampanyası, satışları **artırmıştır** (sadece bu yön iddia ediliyor)

Bu durumda α'nın **tamamı (0.05)**, dağılımın **tek bir ucuna** yerleştirilir bu da tek kuyruklu testi, aynı α değerinde **çift 
kuyrukludan daha "güçlü" (hassas)** yapar çünkü tüm risk payı tek yöne odaklanmıştır.

## Hangisi Ne Zaman Kullanılır?
- **Çift kuyruklu:** Yönden emin değilsek, ya da her iki yöndeki farkı da önemsiyorsak (çoğu gerçek dünya senaryosunda güvenli ve yaygın tercih)
- **Tek kuyruklu:** Sadece belirli bir yöndeki etkiyle ilgileniyorsak VE bu yönü teste BAŞLAMADAN ÖNCE net bir gerekçeyle belirleyebiliyorsak (örn: "bu kampanya sadece satışları artırabilir, azaltamaz" gibi güçlü bir ön bilgi varsa)

## Dikkat Edilmesi Gereken Nokta
Tek kuyruklu test kullanmak, sonucu "daha kolay anlamlı çıkarma" amacıyla **sonradan** seçilmemelidir, bu bilimsel olmayan bir pratiktir. 
(p-hacking'e benzer bir tuzak). Yön, veriye bakmadan ÖNCE, teorik/mantıksal bir gerekçeyle belirlenmelidir.
