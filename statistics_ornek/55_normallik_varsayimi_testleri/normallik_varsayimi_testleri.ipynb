# 55) Normallik Varsayımı Testleri
Birçok hipotez testi (Z-testi, t-testi, ANOVA gibi) **verinin normal dağıldığı varsayımına** dayanır (CLT sayesinde büyük 
örneklemlerde bu varsayım gevşer, ama küçük örneklemlerde önemlidir). Bu konuda, veriye bakıp "gerçekten normal mi?" sorusunu **istatistiksel olarak** nasıl test edeceğimizi öğreniyoruz.

## Neden Sadece Göz Kararı Yetmez?
Konu 13-14'te histogram/skewness/kurtosis ile "görsel olarak" normalliğe bakmıştık. Ama bu **subjektif** bir değerlendirme "bu histogram bana normal gibi görünüyor" demek, kesin bir karar için yeterli değil. Normallik testleri, bu değerlendirmeyi **objektif bir hipotez testine** çevirir.

## En Yaygın Normallik Testi: Shapiro-Wilk Testi
Kendisi de bir hipotez testidir. Aynı H0/H1/p-değeri mantığı burada da işliyor:
- **H0:** Veri normal dağılıma **uyar**
- **H1:** Veri normal dağılıma **uymaz**

**Dikkat bu, alışık olduğumuzdan TERS bir mantık:** Genelde "reddetmek istediğimiz" şey H1 tarafında olurdu (fark var, etki var gibi). Ama burada **normal dağılıma UYMASINI** H0 olarak kuruyoruz — yani "reddedilmemesini istediğimiz" taraf burada H0. Bu tersine dönüş, 
normallik testlerinde herkesi başta şaşırtır, alışmak biraz zaman alabilir.

## Karar Kuralı (Her Zamanki Gibi)
- **p < α (0.05)** → H0 reddedilir → **Veri normal DEĞİLDİR**
- **p ≥ α (0.05)** → H0 reddedilemez → **Veri normal kabul edilebilir**

## Diğer Normallik Testleri (İsimlerini Bilmen Yeterli)
- **Kolmogorov-Smirnov Testi:** Daha büyük örneklemlerde kullanılır
- **Anderson-Darling Testi:** Kuyruklardaki sapmalara daha duyarlıdır

Pratikte en sık **Shapiro-Wilk** kullanılır (özellikle küçük-orta örneklemlerde), bu yüzden odağımız bu olacak.

## Normallik Sağlanmazsa Ne Olur?
Eğer veri normal değilse, iki yol izlenebilir:
1. **Veriyi dönüştürmek** (Konu 13'te gördüğümüz log, karekök gibi dönüşümler — çarpıklığı düzeltmek için)
2. **Non-parametrik test kullanmak** (bunları Non-Parametrik Testler bölümünde detaylı göreceğiz).

```python
from scipy import stats
import numpy as np

normal_veri = np.random.normal(loc=0, scale=1, size=100)
beta = 1/5
exponential_veri = np.random.exponential(scale=beta, size=100)
normal_shapiro = stats.shapiro(x=normal_veri)
expo_shapiro = stats.shapiro(x=exponential_veri)
if (normal_veri_p_value:=normal_shapiro.pvalue) < 0.05:
    print(f'P değeri {round(normal_veri_p_value,5)}, Normal dağılıma uygun veri normal dağılmıyor, H0 reddedilir.')
else:
    print(f"P değeri {round(normal_veri_p_value,5)}, Normal dağılıma uygun veri normal dağılıyor, H0'ı ret edecek yeterli kanıt yok.")

if (expo_veri_p_value:=expo_shapiro.pvalue) < 0.05:
    print(f'P değeri {round(expo_veri_p_value,10)}, Exponential dağılıma uygun veri normal dağılmıyor H0 reddedilir.')
else:
    print(f"P değeri {round(expo_veri_p_value,10)}, Exponential dağılıma uygun veri normal dağılıyor, H0'ı ret edecek yeterli kanıt yok.")
```

### Sonuç
Yukarıda normal dağılıma ve üstel dağılıma uygun oluşturduğumuz iki farklı veri setinin normal dağılıp dağılmadığını test etmek amacıyla oluşturduğumuz hipotezlerde normal dağılıma ait verinin beklenen olarak normal dağıldığını shapiro-wilk testinden elde ettiğimiz p değeri ile gösterdik (p>0.05). Ayrıca üstel dağılıma ait verinin beklenen olarak normal dağılmadığını da shapiro-wilk testi sonucu elde ettiğimiz p değeri ile gösterdik (p<0.05).
