# 49) Hipotez Testinde Tip I ve Tip II Hatalar
Hipotez testinde her zaman **yanlış karar verme riski** vardır çünkü elimizde sadece bir örneklem var, gerçeği (popülasyonun tam durumunu) asla kesin bilemeyiz. İki farklı yanlış karar türü vardır.

## Tip I Hata 
**H₀ aslında DOĞRUYKEN, yanlışlıkla REDDETMEK.** Yani gerçekte hiçbir fark yokken, "fark var" demek — **yanlış alarm.**

*Örnek:* İlaç gerçekten etkisizken, "etkili" sonucuna varmak.

Bu hatanın olasılığı, tam olarak **α** (anlamlılık düzeyi) ile ölçülür — "%5 yanılma riski" tam olarak Tip I Hata riskiydi.

## Tip II Hata (β ile gösterilir)
**H₀ aslında YANLIŞKEN, yanlışlıkla REDDEDEMEMEK.** Yani gerçekte bir fark varken, bunu **kaçırmak** testin gerçek etkiyi **yakalayamaması.**

*Örnek:* İlaç gerçekten etkiliyken, "yeterli kanıt yok" deyip etkisiz sanmak.

Bu hatanın olasılığı **β** ile gösterilir. $1-\beta$ ise **"Test Gücü" (Statistical Power)** olarak adlandırılır (testin, gerçek bir etkiyi yakalama kapasitesi.)

## Özet Tablo
| | H₀ Gerçekte Doğru | H₀ Gerçekte Yanlış |
|---|---|---|
| **H₀'ı Reddettik** | Tip I Hata (α) — Yanlış Alarm | Doğru Karar ✓ |
| **H₀'ı Reddedemedik** | Doğru Karar ✓ | Tip II Hata (β) — Kaçırma |

## İkisi Arasındaki Denge (Trade-off)
Tip I Hata riskini azaltmak için α'yı küçültürsek (örn: 0.05'ten 0.01'e), bu genelde Tip II Hata riskini (β) **artırır** çünkü daha katı bir eşik koyduğun için, gerçek etkileri de daha zor "yakalarız." Bu, hipotez testinin doğasında var olan bir **denge/trade-off**'tur; ikisini de aynı anda sıfıra indiremeyiz (örneklem büyüklüğünü artırmak dışında).

## Hangisi Daha "Kötü"? Bağlama Göre Değişir
- Tıpta: Tip II Hata (gerçekten hasta olan birini "sağlıklı" sanmak) genelde Tip I Hata'dan daha tehlikeli kabul edilir
- Adli sistemde: Tip I Hata (masum birini suçlu bulmak) genelde Tip II'den daha ciddi kabul edilir ("bin suçlu kaçsın, bir masum yanmasın" mantığı)
- Pazarlamada: Bağlama göre değişir. Gereksiz yere kampanyaya devam etmek (Tip I) mi, yoksa gerçekten işe yarayan bir stratejiyi terk etmek (Tip II) mi daha maliyetli, düşünülmesi gerekir.
