# 20) Z-Skor Nedir?
Z-skor, bir değerin ortalamadan kaç standart sapma uzaklıkta olduğunu gösteren standartlaştırılmış bir sayıdır. Bir değerin z skorunu bulmak için, o değeri veri kümesinin ortalamasından çıkarıp yine veri kümesinin standart sapmasına bölerek buluruz.
**Bu değer**
- |Z| = 0 ise değer tam ortalamada
- |Z| = 1 ise değer ortalamadan 1 sapma yukarıda
- |Z| > 2 ise (yani Z, +2'den büyük VEYA -2'den küçükse) **genelde** aykırı değer sayılır (normal dağılımda hem çok yüksek hem çok düşük uç değerler nadir görülür).

**Önemli**: Z skoru farklı ölçeklerdeki verileri ortak paydaya getirmek için biçilmiş kaftandır. Farklı birimden verilerin yorumlanmasını kolaylaştıran istatistiksel bir terimdir.
