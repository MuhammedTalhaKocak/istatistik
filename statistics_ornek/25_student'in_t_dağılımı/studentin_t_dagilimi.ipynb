# 25) Student T Dağılımı Nedir?
Student T Dağılımı, örneklem boyutu küçük olduğunda (n<=30) ve anakütle varyansı bilinmediğinde kullanılan, ortalaması sıfır olan simetrik bir olasılık dağılımıdır. Normal dağılıma göre uç kısımların ihtimalini daha fazla gösterir.

### Serbestlik Derecesi ile İlişkisi
t Dağılımının şekli, serbestlik derecesine (df = n-1) bağlıdır:
- **df küçükken** (örneklem küçük): dağılım daha basık, kuyruklar daha kalın belirsizlik yüksek olur.
- **df büyüdükçe** (örneklem büyüdükçe): t dağılımı giderek **normal dağılıma yaklaşır** çünkü örneklem büyüdükçe, örneklemden hesaplanan 
std sapma popülasyonun gerçek std sapmasına daha çok yaklaşır, belirsizlik azalır.

Bu yüzden n >= 30 civarında artık t dağılımı ile normal (Z) dağılımı arasındaki fark pratik olarak önemsiz hale 
gelir bu da neden büyük örneklemlerde Z-testinin, küçük örneklemlerde ise t-testinin tercih edildiğini açıklar.
