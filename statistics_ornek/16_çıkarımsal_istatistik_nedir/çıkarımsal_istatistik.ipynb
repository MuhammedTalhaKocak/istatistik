# 16) Çıkarımsal İstatistik Nedir?
Çıkarımsal istatistik, küçük bir örneklem verisini kullanarak büyük bir ana kütle (popülasyon) hakkında tahmin yapma, genelleme ve hipotez testi süreçlerini içeren bir veri analizi yöntemidir.

### Çıkarımsal İstatistik ile Betimsel İstatistiğin Karşılaştırılması
- **Betimsel İstatistik**:Sadece eldeki veriyi özetler.
- **Çıkarımsal İstatistik**:Eldeki veriden (örneklem) yola çıkarak daha büyük bir resim (popülasyon) hakkında tahmin yapar.

### Çıkarımsal İstatistikte Temel Kavramlar
- Örneklem (Sample): Üzerinde ölçüm yapılan küçük grup.
- Ana Kütle (Population): Hakkında bilgi edinilmek istenen büyük grup.
- Hipotez Testi: Bir iddianın doğru olup olmadığını matematiksel olarak kontrol etme süreci.
- Güven Aralığı (Confidence Interval): Tahminin ne kadar güvenilir olduğunu gösteren sayı aralığı.
