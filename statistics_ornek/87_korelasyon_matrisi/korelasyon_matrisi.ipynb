# Korelasyon Matrisi Nedir?
Korelasyon Matrisi, birden fazla değişken arasındaki **tüm ikili 
korelasyonları, tek bir tabloda** gösterir — Varyans-Kovaryans 
Matrisi'nin (hatırlarsan) standartlaştırılmış/korelasyon versiyonu.

## Yapısı
$n$ değişken için $n \times n$ boyutunda bir matris:
- **Köşegen:** Her zaman **1** (bir değişkenin kendisiyle korelasyonu 
  her zaman mükemmel)
- **Köşegen dışı:** İkili korelasyon katsayıları (r veya rho)
- **Simetrik:** $Cor(A,B) = Cor(B,A)$ olduğu için, üst üçgen = alt üçgen

## Python'da Kullanımı
```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

reklam = np.random.normal(500, 100, 50)
satis = reklam * 0.8 + np.random.normal(0, 50, 50)
sikayet = np.random.normal(10, 3, 50)
musteri_sayisi = reklam * 0.3 + np.random.normal(0, 20, 50)

df = pd.DataFrame({
    'Reklam': reklam,
    'Satis': satis,
    'Sikayet': sikayet,
    'Musteri_Sayisi': musteri_sayisi
})

# Korelasyon matrisi
korelasyon_matrisi = df.corr(method='pearson')  # method='spearman' de olabilir
print(korelasyon_matrisi)

# Isı haritası (heatmap) ile görselleştirme
plt.figure(figsize=(8,6))
sns.heatmap(korelasyon_matrisi, annot=True, cmap='coolwarm', center=0, vmin=-1, vmax=1)
plt.title("Korelasyon Matrisi")
plt.show()
```

## Heatmap Yorumlama
- **Koyu kırmızı (1'e yakın):** Güçlü pozitif ilişki
- **Koyu mavi (-1'e yakın):** Güçlü negatif ilişki
- **Beyaz/açık renk (0'a yakın):** Zayıf/yok ilişki

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

reklam = np.random.normal(500, 100, 50)
satis = reklam * 0.8 + np.random.normal(0, 50, 50)
sikayet = np.random.normal(10, 3, 50)
musteri_sayisi = reklam * 0.3 + np.random.normal(0, 20, 50)

df = pd.DataFrame({
    'Reklam': reklam,
    'Satis': satis,
    'Sikayet': sikayet,
    'Musteri_Sayisi': musteri_sayisi
})

# Korelasyon matrisi
korelasyon_matrisi = df.corr(method='pearson') 
print(korelasyon_matrisi)

# Isı haritası (heatmap) ile görselleştirme
plt.figure(figsize=(8,6))
sns.heatmap(korelasyon_matrisi, annot=True, cmap='coolwarm', center=0, vmin=-1, vmax=1)
plt.title("Korelasyon Matrisi")
plt.show()
````

Korelasyon matrisini incelediğimizde, satış - reklam (0.877505), müşteri sayısı - reklam (0.859774), müşteri sayısı - satış (0.708744) bu 3 ikili arasında güçlü düzeyde pozitif bir ilişki varken, şikayet - müşteri sayısı (-0.073258), reklam - şikayet (-0.125590) ikilileri arasında ise çok düşük düzeyde negatif bir ilişki vardır. Pozitif düzeydeki ilişkiler, x artarken y de artıyor şeklinde yorumlanabilirken, negatif düzeydeki ilişkiler ise x artarken y azalıyor şeklinde yorumlanabilir. Ayrıca, Reklam ile Satış (0.88) ve Reklam ile Müşteri Sayısı (0.86) arasındaki korelasyonların oldukça yüksek olması dikkat çekicidir. Bu düzeyde yüksek korelasyonlar, ileride bir regresyon modelinde bu değişkenlerin birlikte kullanılması durumunda çoklu doğrusal bağlantı (multicollinearity) sorununa yol açabileceğine işaret etmektedir.
