# 47) Hipotez Testinde Anlamlılık Düzeyi (Significance Level - α)
Konu 48'de şunu sormuştuk: "H₀ doğru olsaydı, gördüğümüz sonucu görme ihtimali ne kadar düşük olurdu, ve bu ne zaman 'yeterince düşük' sayılır?" **Anlamlılık düzeyi (α, alpha)**, işte bu "yeterince düşük" eşiğini **önceden belirlediğimiz** bir sınırdır.

## Tanım
α, H₀'ı **yanlışlıkla reddetme** riskini ne kadar göze aldığımızı gösteren bir eşik değeridir. En yaygın kullanılan değer **α = 0.05**'tir (%5).

## Nasıl Kullanılır?
Test sonucunda bulduğumuz olasılığı (p-değeri, Konu 50'de göreceğiz) α 
ile karşılaştırırız:
- Eğer bulduğumuz olasılık **α'dan küçükse** → H₀'ı reddederiz ("bu kadar 
  nadir bir şey tesadüfen olamaz")
- Eğer bulduğumuz olasılık **α'dan büyükse** → H₀'ı reddedemeyiz

## α = 0.05 Ne Anlama Gelir?
"%5 ihtimalle yanlış bir karar (H₀'ı yanlışlıkla reddetme) vermeyi göze alıyorum" demektir. Yani 100 kere aynı testi tekrarlasak, ve H₀ aslında doğru olsa bile, ortalama 5'inde yanlışlıkla "fark var" sonucuna varabiliriz bu kabul edilen bir risk payıdır.

## Neden Tam %5? Değiştirilebilir mi?
%5, sektörde yaygın bir gelenek/konvansiyondur, matematiksel bir zorunluluk değildir. Bazı alanlarda (örn: tıp, ilaç testleri gibi hata payının çok tehlikeli olduğu yerlerde) α = 0.01 gibi daha katı bir eşik kullanılır. Bazı keşifsel/düşük riskli analizlerde α = 0.10 da kullanılabilir. Önemli olan, **α'yı teste BAŞLAMADAN ÖNCE belirlemek** sonuçları gördükten 
sonra "işimize gelen" bir α seçmek bilimsel olarak yanlıştır (buna "p-hacking" denir, ileride değinebiliriz).
