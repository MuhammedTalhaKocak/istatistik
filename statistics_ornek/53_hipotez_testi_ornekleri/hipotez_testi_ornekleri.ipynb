# 51) Hipotez Testi Örnekleri
Şimdiye kadar öğrendiğimiz kavramları (H0/H1, α, p-değeri) iki kısa senaryo üzerinden, adım adım bir "akış" halinde pekiştirelim. Bu 
konuda henüz spesifik bir test formülü kullanmıyoruz sadece **düşünce sürecini** (hipotez kurma → α belirleme → sonucu yorumlama) tekrarlıyoruz. Formüllerin kendisini ilerleyen konularda (Test İstatistiği, Z/T Testleri) göreceğiz.

## Örnek 1 — E-ticaret Teslimat Süresi
Bir e-ticaret şirketi, ortalama teslimat süresinin **3 gün** olduğunu iddia ediyor. Biz şüpheliyiz, gerçekte daha uzun sürdüğünü düşünüyoruz.

- **H0:** Ortalama teslimat süresi 3 gündür (iddia doğru)
- **H1:** Ortalama teslimat süresi 3 günden **fazladır**
- **Kuyruk yapısı:** Tek kuyruklu (Konu 48, sadece "fazla" yönünü iddia ediyoruz)
- **α:** 0.05 olarak belirledik (teste başlamadan önce)

Diyelim ki 50 siparişten örneklem topladık, ortalama teslimat süresi 3.8 gün çıktı, ve hesapladığın p-değeri **0.02** çıktı.

**Karar:** p (0.02) < α (0.05) → H0'ı reddederiz. Sonuç: teslimat süresi gerçekten 3 günden fazla, istatistiksel olarak anlamlı bir şekilde.

## Örnek 2 — Yeni Web Sitesi Tasarımı
Bir şirket, yeni web sitesi tasarımının dönüşüm oranını **değiştirmediğini** düşünüyor (temkinli bir varsayım).

- **H0:** Yeni tasarımın dönüşüm oranı, eski tasarımla **aynıdır**
- **H1:** Yeni tasarımın dönüşüm oranı eski tasarımdan **farklıdır**
- **Kuyruk yapısı:** Çift kuyruklu (yön belirtmiyoruz — hem artış hem azalış "farklı" sayılır)
- **α:** 0.05

Diyelim ki testi yaptık, p-değeri **0.34** çıktı.

**Karar:** p (0.34) ≥ α (0.05) → H0'ı reddedemedik. Sonuç: elimizdeki veri, yeni tasarımın gerçekten bir fark yarattığını kanıtlayacak kadar güçlü değil (dikkat: "tasarım kesinlikle etkisiz" demiyoruz, sadece "yeterli kanıt yok" diyoruz).

## İki Örnekten Çıkan Ortak Ders
Her iki örnekte de akış **aynı 4 adımı** izledi:
1. H0 ve H1'i teste başlamadan önce net şekilde kurduk
2. α'yı önceden belirledik (sonuçları görüp α'yı değiştirmek yanlıştır)
3. Veriyi toplayıp, p-değerini hesapladık
4. p'yi α ile karşılaştırıp karar verdik, ve sonucu doğru yorumladık (reddetme ≠ kesin doğrulama)
