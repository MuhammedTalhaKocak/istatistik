## 27) Normal Dağılım Ortalama Güven Aralığı

Normal dağılım ortalama güven aralığı, Popülasyonun standart sapması (σ) biliniyorsa VEYA örneklem büyüklüğü yeterince büyükse (n≥30,merkezi limit teoremi sayesinde), ortalama için güven aralığı Z dağılımı kullanılarak hesaplanır. 

Formül:$$ \bar{x} \pm Z_{\alpha/2} \cdot \frac{s}{\sqrt{n}} $$

% x güven düzeyi için formüldeki alfa değeri % 100-x'e eşit olur. Bu alfa değeri bizim yanılma payımızı ifade eder, alfa/2 kullanmamızın sebebi ise güven aralığının hem alt hem de üst sınırdan oluşmasıdır. 

### Kritik Yorumlama Kısmı
%95 güven aralığı" demek, "gerçek popülasyon ortalamasının bu aralıkta olma olasılığı %95'tir" demek değildir (popülasyon ortalaması sabit bir sayı, olasılıksal değil). Doğru yorum bu şekilde olmalıdır: "Eğer bu süreci (örneklem çekip aralık hesaplama) 100 kere tekrarlasaydık, bu aralıkların yaklaşık 95'i gerçek popülasyon ortalamasını içerirdi.
​

```python
# Uygulama
import numpy as np
from scipy import stats
```

```python
# öncelikle elle güven aralığını hesaplayalım ardından stats modülünden norm.interval fonksiyonunu kullanacağız.
veri = np.random.normal(loc=80, scale=7, size=30) # loc = ortalama, scale = standart sapma, size = gözlem sayısı
print(veri)
```

```python
ortalama = np.mean(veri)
se = np.std(veri, ddof=1) / (len(veri) ** 0.5)  # Standart Error (SE)
# bize gereken şey alt sınırın solunda kalan alan ve üst sınırın sağında kalan alana ait z skorlar. bunun için ppf fonksiyonunu kullanalım.
z_alt = stats.norm.ppf(0.025)
z_ust = stats.norm.ppf(0.975)
guven_araligi_alt_sinir = ortalama + (z_alt * se)
guven_araligi_ust_sinir = ortalama + (z_ust * se)
print(f'%95 Güven aralığı {guven_araligi_alt_sinir} ile {guven_araligi_ust_sinir} arasındadır.')
```

```python
# hazır fonksiyon ile aynı soruyu çözelim.
alt_sinir, ust_sinir = stats.norm.interval(confidence=0.95, loc=ortalama, scale=se) # interval fonksiyonuna güven düzeyi, ortalama (loc) ve scale (standart hatayı) parametre olarak gönderiyoruz.
print(alt_sinir, ust_sinir)
```
