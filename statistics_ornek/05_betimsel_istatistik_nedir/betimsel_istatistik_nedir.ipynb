# 5) Betimsel İstatistik Nedir?
Betimsel İstatistik, istatistiğin verilerin toplanması, düzenlenmesi ve özetlenmesi gibi sadece tanımlayıcı açıdan incelemeye dayanan ve herhangi bir olasılıksal sürece dayanmayan bir alt dalıdır.
## Betimsel İstatistiğin 3 Temel Boyutu
1. **Merkezi Eğilim**: Veri nerede yoğunlaşıyor? → Ortalama, Medyan, Mod
2. **Yayılım/Dağılım**: Veri ne kadar yayılmış? → Ranj, Varyans, Standart Sapma
3. **Şekil**: Dağılım simetrik mi, çarpık mı? → Çarpıklık, Basıklık

Bu üçü **birlikte** kullanılmalı, sadece merkezi eğilime bakmak yanıltıcı olabilir (örneğin ortalama tek başına, uç değerlerin etkisini gizleyebilir).
