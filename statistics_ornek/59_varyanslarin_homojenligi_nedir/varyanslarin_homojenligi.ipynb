# 59) Varyansların Homojenliği Nedir?
İki grubu karşılaştırırken (Bağımsız İki Örneklem T Testi), formülün doğru çalışması için önemli bir varsayım vardır: **iki grubun 
varyanslarının birbirine yakın (homojen) olması.** Bu varsayımın sağlanıp sağlanmadığını test etmeden `equal_var=True` (Pooled) kullanmak, riskli bir varsayımdır.

## Neden Önemli?
Konu 31'de gördüğümüz gibi, Pooled yöntem, iki grubun varyansını "havuzlayarak" tek bir ortak varyans hesaplar ama eğer gerçekte iki grubun varyansı çok farklıysa, bu havuzlama **yanıltıcı** bir standart hata üretir, bu da yanlış p-değerlerine yol açabilir.

## Hipotezler (Homojenlik Testinin Kendisi de Bir Hipotez Testidir)
- **H0:** İki grubun varyansları eşittir (homojendir)
- **H1:** İki grubun varyansları eşit değildir

(Normallik testindeki gibi yine "ters" bir mantık reddetmek istemediğimiz taraf H0.)

## Karar Kuralı
- p < α → H0 reddedilir → Varyanslar **eşit değil** → `equal_var=False` (Welch) kullanılmalı
- p ≥ α → H0 reddedilemedi → Varyanslar **eşit kabul edilebilir** → `equal_var=True` (Pooled) kullanılabilir

## En Yaygın İki Test
Bir sonraki konumuzda (Bartlett ve Levene Testleri) bu iki testi detaylı göreceğiz şimdilik sadece varyans homojenliğinin **neden** kontrol edilmesi gerektiğini anlamaya çalışıyoruz.
