# 65) ANOVA (Analysis of Variance) Nedir?
ANOVA, **ikiden fazla** grubun ortalamalarını, **tek bir testte, aynı anda** karşılaştırmamızı sağlayan bir hipotez testidir. T testi sadece 2 grup karşılaştırabilirken, ANOVA bu sınırı kaldırır — 3, 4, 10 grup fark etmez, hepsini bir arada test eder.

## Neden İkili T Testleri Yerine ANOVA?
Konu (FamilyWise Error Rate)'de gördüğümüz gibi, grupları ikişer ikişer T testiyle karşılaştırmak, **Alpha Inflation** riskini doğurur — çok sayıda karşılaştırma, toplam Tip I Hata riskini şişirir. ANOVA, tüm grupları TEK BİR testte değerlendirerek, bu riski **kontrol altında** tutar — sadece **bir kere** α=0.05 riski almamızı sağlar, kaç grup olursa olsun.

## Hipotezler
- **H0:** Tüm grupların ortalamaları eşittir ($\mu_1 = \mu_2 = \mu_3 = ...$)
- **H1:** En az bir grup, diğerlerinden farklıdır (hangi grubun farklı  olduğunu ANOVA'nın kendisi söylemez, sadece "en az biri farklı" der — "hangisi" sorusunun cevabını Post-Hoc testler verecek, Konu 67)

## İsmindeki Paradoks — "Varyans Analizi" Ama Ortalamaları Karşılaştırıyor?
ANOVA'nın adı "Varyans Analizi" ama aslında **ortalamaları** karşılaştırır kafa karıştırıcı gelebilir. Mantığı şöyle: ANOVA, **iki farklı varyans kaynağını** karşılaştırarak çalışır:
1. **Gruplar arası varyans (Between-Group):** Grup ortalamalarının birbirinden ne kadar farklı olduğu
2. **Grup içi varyans (Within-Group):** Her grubun kendi içindeki doğal dağınıklığı (rastgele varyasyon)

Eğer "gruplar arası" varyans, "grup içi" varyansa göre **çok büyükse**, bu "gruplar arasında gerçek bir fark var" demektir — sadece rastgele varyasyon değil.

## Test İstatistiği: F-İstatistiği
$$F = \frac{\text{Gruplar Arası Varyans}}{\text{Grup İçi Varyans}}$$

F büyükse (gruplar arası fark, grup içi dağınıklığa göre baskınsa) → H0 reddedilir. F küçükse (gruplar arası fark, grup içi rastgele varyasyondan pek farklı değilse) → H0 reddedilemez.

## ANOVA'nın Varsayımları (Önceden Kontrol Edilmeli)
1. **Normallik:** Her grup normal dağılıma yakın olmalı (Shapiro-Wilk, Konu 55)
2. **Varyans Homojenliği:** Grupların varyansları birbirine yakın olmalı (Levene, Konu 60)
3. **Bağımsızlık:** Gruplar birbirinden bağımsız olmalı

## Türleri (İlerleyen Konularda Detaylı Göreceğiz)
- **One-Way ANOVA (Konu 66):** Tek bir kategorik değişkene göre gruplama (örn: sadece "Bölge")
- **Two-Way ANOVA (Konu 68):** İki kategorik değişkene göre gruplama (örn: "Bölge" VE "Segment" birlikte)
