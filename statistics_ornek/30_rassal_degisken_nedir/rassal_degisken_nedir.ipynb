# 30) Rassal Değişken Nedir?
Rassal Değişken (Random Variable), sonucu şansa/rastgeleliğe bağlı olan bir 
deneyin çıktısına sayısal bir değer atayan değişkendir. Deneyin kendisi 
rastgele olsa da, rassal değişken bu rastgeleliği **sayılarla ifade 
edilebilir** hale getirir bu sayede olasılık ve istatistik araçlarını 
üzerinde kullanabiliriz.

### İki Tür Rassal Değişken

**1. Kesikli (Discrete) Rassal Değişken**
Sayılabilir, belirli ve ayrık değerler alır. İki değer arasında "ara" bir 
değer olamaz.

- *Örnek:* Bir zar atıldığında gelen sayı → sadece {1, 2, 3, 4, 5, 6} 
  değerlerinden birini alabilir, 3.5 gibi bir sonuç imkansızdır.
- *Örnek:* Bir mağazaya bir günde gelen müşteri sayısı → 0, 1, 2, 3... 
  şeklinde sayılabilir, 4.7 müşteri diye bir şey olamaz.

**2. Sürekli (Continuous) Rassal Değişken**
Belirli bir aralıkta **sonsuz sayıda** değer alabilir, iki değer arasında 
her zaman bir ara değer bulunabilir.

- *Örnek:* Bir kişinin boyu → 172 cm, 172.3 cm, 172.34 cm... teorik olarak 
  ölçüm hassasiyeti arttıkça sonsuz farklı değer mümkündür.
- *Örnek:* Bir siparişin teslimat süresi (saat cinsinden) → 2 saat, 2.5 
  saat, 2.53 saat gibi kesintisiz bir aralıkta herhangi bir değer olabilir.

### Neden Önemli?
Rassal değişken kavramı, olasılık teorisinin temel yapı taşıdır. Bundan 
sonra göreceğimiz **Beklenen Değer**, **Olasılık Dağılımları** (Bernoulli, 
Binom, Poisson, Normal gibi) hep bir rassal değişkenin **hangi değerleri 
ne olasılıkla aldığını** tanımlamaya çalışır. Bir değişkenin kesikli mi 
sürekli mi olduğunu bilmek, hangi dağılım/formülün uygulanacağını 
belirleyen ilk adımdır örneğin Binom dağılımı sadece kesikli değişkenler 
için, Normal dağılım ise sürekli değişkenler için tanımlıdır.

### Küçük Bir Ayrım Notu
Rassal değişken, Konu 3'te gördüğümüz "nicel değişken" kavramına benzer 
ama tam olarak aynı değildir rassal değişken vurgusu, değerin **henüz 
gerçekleşmemiş, olasılığa bağlı** olmasında; nicel değişken ise zaten 
**gözlemlenmiş, elimizdeki veri setinde kayıtlı** bir değeri ifade eder. 
Yani zar atmadan önceki sonuç bir rassal değişkendir; zar atıldıktan ve 
"5 geldi" diye kaydedildikten sonra bu artık gözlemlenmiş bir veri 
noktasıdır.

```python
import numpy as np

# Adil bir zarın olası değerleri (kesikli rassal değişken)
zar_degerleri = [1, 2, 3, 4, 5, 6]
olasiliklar = [1/6] * 6  # Her yüzün gelme olasılığı eşit

for deger, olasilik in zip(zar_degerleri, olasiliklar):
    print(f"X = {deger} olma olasılığı: {olasilik:.4f}")

# Simülasyon: zarı 10000 kere atalım, sonuçların dağılımına bakalım
np.random.seed(42)
zar_atislari = np.random.choice(zar_degerleri, size=10000)

# Her değerin kaç kere geldiğini (deneysel olasılığı) hesaplayalım
degerler, sayilar = np.unique(zar_atislari, return_counts=True)
for d, s in zip(degerler, sayilar):
    print(f"X = {d}: {s} kere geldi, deneysel olasılık = {s/10000:.4f}")
```

Bir zara ait gelebilecek olası sonuçları zar_degerleri listesine ekledikten sonra bu listeden zar_atislari listesine random.choice fonksiyonu yardımıyla 10000 gözlemli bir atış çekme işlemi uygulayıp for döngüsüyle bu listedeki her bir benzersiz elemanın frekansını yazdırdık. Sonuç olarak örneklemin yeterince büyük olması sebebiyle deneysel olasılıklar teorik olasılığa oldukça yakın çıktı.
