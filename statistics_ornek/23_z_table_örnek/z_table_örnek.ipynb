```python
from scipy import stats
import numpy as np
# Elimizdeki rastgele Z skorlarının Z tablosundaki yerlerini (bu z skorlarının solunda kalan alanları) bulalım.
z_skorlari = [1.96, -1.96, 0, 1]
for z_skor in z_skorlari:
    print(f'Z-Skor {z_skor} için bu skorun solunda kalan kümülatif alanı: {stats.norm.cdf(x=z_skor)}')
```

Yukarıdaki Z skorlarına ait sol taraflarında kalan kümülatif alanı bulmak için stats modülünden normal dağılıma ait cdf (kümülatif alan hesaplama) fonksiyonunu kullandık. burada Z Skor = 1.96 ve -1.96 değerlerine karşılık gelen sol taraflarındaki kümülatif olasılık alanlarını çıkardığımızda %95 değerini elde ediyoruz bu da %95 güven aralığı kavramının nerden geldiğinin bir kanıtı aslında.
