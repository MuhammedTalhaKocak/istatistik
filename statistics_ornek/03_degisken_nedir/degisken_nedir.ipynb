# 3) Değişken Nedir?
Değişken, bir veri setindeki her bir kolon/özelliğe verilen isimdir. Gözlemden gözleme farklı değerler alır.
## Değişken Türleri
- **Nicel Değişken**: Sayısal değişkenlerdir. Bu değişkenler üzerinde anlamlı olarak matematiksel işlemler (ortalama, standart sapma vs.) yapılabilmektedir ancak sayısal olarak gözüken her değişken nicel değişken değildir. Örneğin: TC Kimlik No (Sayısal olarak gözükse bile insanların TC Kimlik Numaralarının) ortalaması/standart sapması alınması mantıksızdır.
- **Nitel Değişken**: Sayısal olmayan değişkenlerdir. Grup/etiket belirtirler. Örneğin: Cinsiyet, Kadın-Erkek şeklinde grup belirtir.
