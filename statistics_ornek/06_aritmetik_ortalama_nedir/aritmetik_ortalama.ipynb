# 6) Aritmetik Ortalama Nedir?
Aritmetik ortalama, bir sayı dizisindeki elemanların toplamının eleman sayısına bölünmesi ile hesaplanan bir merkezi eğilim ölçüsüdür. Aritmetik ortalama uç değerlere (outlier) karşı çok hassastır ve bu değerlerden doğrudan etkilenir. Bir veri kümesindeki çok büyük veya çok küçük tek bir sayı bile, tüm aritmetik ortalama sonucunu gerçek durumu gizleyecek kadar yukarı veya aşağı çekebilir. 

```python
import numpy as np
gelirler = np.array([30000, 40000, 40000, 50000, 1000000]) 
gelir_ortalamasi = np.mean(gelirler) # ortalama hesaplamada NumPy kütüphanesinden mean fonksiyonu kullanıyoruz, içine sayı dizimizi parametre olarak gönderiyoruz.
print(f"Bütün Gelirlerin ortalaması: {gelir_ortalamasi}")
print(f"Yüksek gelirli arkadaş harici diğer kişilerin gelir ortalaması : {np.mean(gelirler, where=(gelirler != 1000000))}")
```

#### Bu arkadaş grubundaki çok yüksek gelirli kişiyi dahil ettiğimizde bu grubun gelir ortalaması 232000 iken, bu kişiyi gruptan çıkardığımızda geriye kalan kişilerin maaş ortalaması 40000 olmuştur. Bu da aşırı uç bir değerin ortalamayı ne kadar yukarı çektiğinin bir başka deyişle ortalamanın aşırı uç değerler yüzünden veriyi doğru bir biçimde yansıtamadığının bir kanıtıdır.
