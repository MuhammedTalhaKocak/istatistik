# 48) Hipotez Testi Nedir?
Hipotez Testi, bir iddiayı (varsayımı) **veriye dayanarak matematiksel olarak sınama** sürecidir. "Bu iki grup arasında gerçekten bir fark var mı, yoksa gördüğümüz fark sadece rastgele şans mı?" sorusuna istatistiksel olarak cevap vermeye çalışırız.

## Bağlantı Noktası
Bu bölümdeki her şey, Çıkarımsal İstatistik bölümünde (Konu 16-30) öğrendiklerimizin **doğrudan uygulamasıdır.** Z-Score, t-dağılımı, standart hata — hepsini burada kullanacağız, ama artık "güven aralığı bulma" değil, **"karar verme"** amacıyla.

## Temel Mantık (Büyük Resim)
1. Bir iddia kurarız (örneğin: "yeni pazarlama stratejisi satışları artırmıyor" buna **Boş Hipotez** denir, Konu 46'da göreceğiz)
2. Elimizdeki veriye bakarız
3. Kendimize şunu sorarız: **"Bu iddia gerçekten doğru olsaydı, elimizdeki bu veriyi (ya da daha uç bir sonucu) görme ihtimalimiz ne kadar düşük olurdu?"**
4. Eğer bu ihtimal **çok düşükse**, iddiayı (Boş Hipotezi) reddederiz çünkü "bu kadar nadir bir şey tesadüfen olmuş olamaz, demek ki gerçek bir fark var" deriz

## Neden Önemli?
- **A/B testleri:** "Yeni buton rengi tıklama oranını artırıyor mu?"
- **Pazarlama kampanyası etkinliği:** "Bu kampanya gerçekten satışları artırdı mı, yoksa mevsimsel bir dalgalanma mı?"
- **Ürün/süreç değişiklikleri:** "Yeni bir teslimat süreci, müşteri memnuniyetini gerçekten artırdı mı?"

Bu tür sorularda "gözle bakınca fark var gibi duruyor" demek yeterli değildir. Hipotez testi, bu farkın **istatistiksel olarak anlamlı** olup olmadığını kesin bir çerçeveyle ölçer.
