# 17) Merkezi Limit Teoremi Nedir?
Merkezi Limit Teoremi, bir veri kümesinin (popülasyon) dağılımı ne olursa olsun bu popülasyondan yeterli büyüklükte (en az 30 kişi olarak kabul edilen) alınan örneklemlerin dağılımlarının normal dağılıma yakınsayacağını söyler. Bu söylem, örneklem ortalamalarının normal olduğu varsayımlarıyla çalışan test istatistikleri için kritiktir.
