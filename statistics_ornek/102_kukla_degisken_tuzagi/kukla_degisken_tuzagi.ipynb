# Kukla Değişken Tuzağı (Dummy Variable Trap)
Bir önceki konuda kendi kendine keşfettiğin sorunu şimdi resmi olarak 
ele alıyoruz: N kategorili bir değişkeni One-Hot Encoding ile kodlarken, 
**N sütunun hepsini** tutarsak, bu sütunlar arasında **mükemmel bir 
doğrusal bağımlılık** oluşur — bu, ileride Regresyon'da öğreneceğimiz "multicollinearity" sorununun en 
uç/kesin halidir.

## Neden Sorun Yaratır?
Bazı algoritmalar, matematiksel olarak bu **mükemmel bağımlılığı çözemez** — model kurulamaz hale gelir ya da katsayılar anlamsız/kararsız çıkar.

## Çözüm: Bir Kategoriyi "Referans" Olarak Bırakmak
N kategori varsa, sadece **N-1 sütun** tutulur. Çıkarılan kategori 
"referans/temel kategori" olur — diğer tüm sütunlar, o referansa göre 
yorumlanır (0,0,0 → otomatik olarak "referans kategori" demektir).

```python
onehot_duzeltilmis = pd.get_dummies(df['Renk'], drop_first=True)
print(onehot_duzeltilmis)
```

Bu durumda, örneğin "Kırmızı" sütunu çıkarılır (alfabetik ilk kategori 
genelde referans olur) — geriye sadece "Mavi" ve "Yeşil" sütunları 
kalır. Bir satırda ikisi de 0 ise, bu **otomatik olarak Kırmızı** demektir.

## Ne Zaman Bu Tuzağa Dikkat Etmeli, Ne Zaman Önemsiz?
- **Doğrusal Regresyon, Lojistik Regresyon gibi modellerde:** Kritik, mutlaka `drop_first=True` kullanılmalı.
- **Ağaç tabanlı modeller (Random Forest, XGBoost gibi):** Bu tuzaktan etkilenmezler, N sütunun hepsi tutulabilir

````python
import pandas as pd

df = pd.DataFrame({'Renk': ['Kırmızı', 'Mavi', 'Yeşil', 'Kırmızı', 'Mavi', 'Yeşil']})

onehot = pd.get_dummies(df['Renk'], drop_first = True)
print("One-Hot Encoding (tüm kategoriler):")
print(onehot)
````

### Sonuç
Bir önceki konuda değindiğimiz **kukla değişken tuzağı** sorunun oluşabilmesinin önüne geçtik. Şu anda gereksiz bir sütun daha türemedi **drop_first = True** ifadesiyle.
