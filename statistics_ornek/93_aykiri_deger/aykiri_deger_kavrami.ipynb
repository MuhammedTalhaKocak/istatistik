# Aykırı Değer (Outlier) Nedir?
Aykırı Değer, bir veri setindeki diğer gözlemlerden **anormal derecede uzak** olan gözlemdir.
## Neden Önemli?
Aykırı değerler:
- Ortalama, standart sapma gibi istatistikleri çarpıtabilir.
- Pearson korelasyonunu ciddi şekilde yanıltabilir.
- Hipotez testlerinin sonuçlarını etkileyebilir
- Makine Öğrenmesi modellerinde ciddi performans sorunlarına yol 
  açabilir

## Aykırı Değerin İki Kaynağı
1. **Gerçek ama nadir bir durum:** Örneğin gerçekten çok zengin bir 
   müşterinin devasa bir alışverişi — bu "hatalı" değil, gerçek ve 
   nadir bir olay
2. **Veri hatası:** Yanlış veri girişi (örn: yaş=999 gibi), sensör 
   arızası, birim hatası (TL yerine kuruş girilmesi gibi)

Bu ayrım kritik çünkü **çözüm stratejisi farklı olmalı** — gerçek bir 
nadir olayı silmek, değerli bilgi kaybına yol açabilir; ama bir veri 
hatasını tutmak, analizini tamamen bozabilir.

## Aykırı Değer Türleri
- **Tek Değişkenli (Univariate):** Tek bir değişkende aşırı uç (örn: 
  Yaş=150)
- **Çok Değişkenli (Multivariate):** Tek başına her değişken normal görünse de, **kombinasyon halinde** anormal olan gözlem (örn: 25 
yaşında ama emekli olduğunu belirten bir kayıt — yaş normal, emeklilik durumu normal, ama BİRLİKTE anormal)

## Genel Yaklaşım
Aykırı değer tespiti genelde şu sırayı izler:
1. **Görsel inceleme:** Box plot, scatter plot (gözle bakmak her zaman ilk adım)
2. **İstatistiksel tespit:** Z-Score, IQR yöntemi
3. **Karar:** Sil, dönüştür, baskıla (cap), ya da özel bir kategori olarak işaretle.
