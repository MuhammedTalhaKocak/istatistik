# 9) Kartil Nedir?
Kartil, Türkçe'de dördebölen olarak da adlandırılan, küçükten büyüğe sıralanmış bir veri kümesini 4 eşit parçaya ayıran üç temel değerden her biridir.
### Kartil Çeşitleri
- **Birinci Kartil (Q₁ - Alt Kartil)**:Veri setinin ilk %25'lik dilimini belirleyen, alt grubun medyanıdır.
- **İkinci Kartil (Q₂ - Medyan)**:Veri setini tam ortadan ikiye bölen ve verilerin %50'sini temsil eden ortanca değerdir.
- **Üçüncü Kartil (Q₃ - Üst Kartil)**: Veri setinin %75'lik dilimini sınırlayan, üst grubun medyanıdır.
- **Dördüncü Kartil(Q₄ - Maksimum Değer)**: Verilerin %100'ü bu değerin altındadır (En büyük veri)

### IQR (Kartil Aralığı) Kavramı Nedir?
- Kartil, Üçüncü kartil ile birinci kartil arasındaki farktır (Q₃ - Q₁).
- Veri setinin orta %50'lik kısmının yayılımını ve değişkenliğini gösterir.
- Aşırı uç değerlerden (aykırı değerlerden) etkilenmeyen güvenilir bir yayılım ölçüsüdür.

```python
import numpy as np
satislar = [12, 45, 22, 89, 34, 15, 67, 91, 8, 53, 29, 76]
Q1 = np.quantile(satislar, 0.25) # Kartil hesaplaması için numpy kütüphanesinin quantile fonksiyonunu kullanıyoruz. Bu fonksiyon veri kümesinin belirli bir yüzdelik dilimini hesaplar. 
Q2 = np.quantile(satislar, 0.50)
Q3 = np.quantile(satislar, 0.75)
IQR = Q3 - Q1
print(f"İlk Kartil (Q1): {Q1}")
print(f"İkinci Kartil (Q2): {Q2}, Medyan:{np.median(satislar)}") # burada 2.Kartil ile medyanın aynı çıktığını göreceğiz.
print(f"Üçüncü Kartil (Q3): {Q3}")
print(f"Çeyrekler Arası Açıklık (IQR): {IQR}")
print(f"Minimum Değer: {np.min(satislar)}, Maksimum Değer: {np.max(satislar)}")
```

### Sonuç

belirgin bir aykırı değer olmadan da verinin genelinde doğal bir 
dağınıklık olduğunu gösterir.
