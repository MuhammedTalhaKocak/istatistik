# 50) Hipotez Testinde P Değeri Nedir?

P-değeri, **H₀'ın doğru olduğu varsayımı altında**, elimizdeki veriyi (ya da daha "uç"/aşırı bir sonucu) **görme olasılığıdır.**
Basitçe: "Eğer gerçekte hiçbir fark yoksa (H₀ doğruysa), benim bulduğum bu sonucu (ya da daha çarpıcısını) tesadüfen görme ihtimalim ne kadar?"

## Karar Kuralı
P-değerini, önceden belirlediğimiz α ile karşılaştırırız:
- **p < α** → H₀'ı reddet (sonuç istatistiksel olarak **anlamlı**)
- **p ≥ α** → H₀'ı reddedemedik (sonuç istatistiksel olarak **anlamlı değil**)

## Örnek (Parayla, Sohbetimizde Kullandığımız)
Adil bir para 10 kere atıldı, 10'u da yazı geldi. Bu senaryoda p-değeri yaklaşık **0.001** çıkar (H₀ doğruysa - yani para adilse - 10/10 yazı gelme olasılığı). Bu, α=0.05'ten çok daha küçük olduğu için H₀'ı reddederiz, "para hileli" sonucuna varırız.

## P-Değeri Hakkında Sık Yapılan Yanlış Yorumlar (DİKKAT)
1. **"p-değeri, H₀'ın doğru olma olasılığıdır" — YANLIŞ.** P-değeri, "H₀ DOĞRU OLSAYDI bu veriyi görme olasılığı" demektir, "H₀'ın kendisinin doğru olma olasılığı" değildir. Bu ayrım çok inceliklidir ama kritiktir.
2. **"p < 0.05 ise etki BÜYÜKTÜR" — YANLIŞ.** P-değeri, etkinin **büyüklüğünü** değil, sadece **ne kadar "şaşırtıcı/nadir"** olduğunu gösterir. Çok küçük ama gerçek bir etki, yeterince büyük örneklemle çok küçük bir p-değeri verebilir (bunu ileride Effect Size konusunda 
göreceğiz).
3. **"p > 0.05 ise H₀ kesinlikle doğrudur" — YANLIŞ.** Konu 46'da (Boş/Alternatif Hipotezler) değindiğimiz gibi, bu sadece "H₀'ı 
reddetmek için yeterli kanıt yok" demektir, "H₀ doğrudur" demek değildir.

## Neden Önemli?
P-değeri, hipotez testinin **sonucunu iletmenin standart yoludur** bilimsel makalelerde, A/B test raporlarında, iş sunumlarında sürekli 
karşımıza çıkacak. Doğru yorumlamak, yanlış iş kararları vermemek için kritik.
